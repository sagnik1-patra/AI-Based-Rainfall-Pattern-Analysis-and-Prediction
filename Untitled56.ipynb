```python
import os
import json
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

import joblib
import yaml

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.callbacks import EarlyStopping


warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# ======================================================================
# 2. PROJECT PATHS
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Rainfall Pattern Analysis and Prediction"
)

DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_june_2019_to_may_2020.csv"
)


# ======================================================================
# 3. OUTPUT PATHS
# ======================================================================

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_prediction_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_preprocessor.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_metadata.json"
)


ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "accuracy_graph.png"
)

HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "heatmap.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "comparison_graph.png"
)

RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "result.csv"
)

RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "result_graph.png"
)

PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "prediction.csv"
)

PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "prediction_graph.png"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ======================================================================
# 4. CHECK DATASET
# ======================================================================

print("=" * 75)
print("AI-BASED RAINFALL PATTERN ANALYSIS AND PREDICTION")
print("=" * 75)


if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )


print("\nDataset found successfully.")

print(
    "\nDataset Path:\n",
    DATASET_PATH
)


# ======================================================================
# 5. LOAD DATASET
# ======================================================================

df = pd.read_csv(
    DATASET_PATH
)


# Clean column names
df.columns = (
    df.columns
    .str.strip()
)


print("\n" + "=" * 75)
print("DATASET")
print("=" * 75)

print(
    "\nShape:",
    df.shape
)

print(
    "\nColumns:",
    df.columns.tolist()
)

print(
    "\nData:"
)

print(
    df.to_string(
        index=False
    )
)


# ======================================================================
# 6. IDENTIFY COLUMNS
# ======================================================================

description_column = None
rainfall_column = None


for column in df.columns:

    column_lower = column.lower()

    if "description" in column_lower:

        description_column = column

    if "rainfall" in column_lower:

        rainfall_column = column


# Fallback for first and second columns
if description_column is None:

    description_column = df.columns[0]


if rainfall_column is None:

    numeric_candidates = []

    for column in df.columns:

        converted = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        if converted.notna().sum() > 0:

            numeric_candidates.append(
                column
            )

    if len(numeric_candidates) > 0:

        rainfall_column = numeric_candidates[0]


if rainfall_column is None:

    raise ValueError(
        "Could not identify rainfall column."
    )


print(
    "\nDescription Column:",
    description_column
)

print(
    "Rainfall Column:",
    rainfall_column
)


# ======================================================================
# 7. CLEAN DATA
# ======================================================================

df[description_column] = (
    df[description_column]
    .astype(str)
    .str.strip()
)


df[rainfall_column] = pd.to_numeric(
    df[rainfall_column],
    errors="coerce"
)


df = df.dropna(
    subset=[
        rainfall_column
    ]
).reset_index(
    drop=True
)


# ======================================================================
# 8. GET NORMAL AND ACTUAL RAINFALL
# ======================================================================

normal_rows = df[
    df[description_column]
    .str.lower()
    .str.contains(
        "normal",
        na=False
    )
]


actual_rows = df[
    df[description_column]
    .str.lower()
    .str.contains(
        "actual",
        na=False
    )
]


if len(normal_rows) == 0:

    raise ValueError(
        "Normal Rainfall row not found."
    )


if len(actual_rows) == 0:

    raise ValueError(
        "Actual Rainfall row not found."
    )


normal_rainfall = float(
    normal_rows[
        rainfall_column
    ].iloc[0]
)


actual_rainfall = float(
    actual_rows[
        rainfall_column
    ].iloc[0]
)


# ======================================================================
# 9. BASIC RAINFALL ANALYSIS
# ======================================================================

difference = (
    actual_rainfall
    -
    normal_rainfall
)


percentage_deviation = (
    difference
    /
    normal_rainfall
) * 100


rainfall_ratio = (
    actual_rainfall
    /
    normal_rainfall
)


if percentage_deviation > 0:

    rainfall_pattern = "Above Normal"

elif percentage_deviation < 0:

    rainfall_pattern = "Below Normal"

else:

    rainfall_pattern = "Normal"


print("\n" + "=" * 75)
print("RAINFALL ANALYSIS")
print("=" * 75)


print(
    f"\nNormal Rainfall     : {normal_rainfall:.2f} mm"
)

print(
    f"Actual Rainfall     : {actual_rainfall:.2f} mm"
)

print(
    f"Difference          : {difference:.2f} mm"
)

print(
    f"Percentage Deviation: {percentage_deviation:.4f}%"
)

print(
    f"Rainfall Ratio      : {rainfall_ratio:.6f}"
)

print(
    f"Rainfall Pattern    : {rainfall_pattern}"
)


# ======================================================================
# 10. CREATE TRAINING DATA
# ======================================================================
#
# IMPORTANT:
# The supplied CSV contains only one normal/actual aggregate pair.
#
# Therefore, a local synthetic training set is generated around the
# observed rainfall relationship.
#
# This makes it possible to demonstrate the complete ML pipeline,
# graphs, prediction, serialization and evaluation.
#
# It should NOT be interpreted as a production rainfall forecasting
# model.
# ======================================================================

TOTAL_SAMPLES = 500


normal_values = np.linspace(
    normal_rainfall * 0.60,
    normal_rainfall * 1.40,
    TOTAL_SAMPLES
)


# Small deterministic variation is introduced so that the model is not
# simply trained against a perfectly identical linear relationship.

variation = (
    1
    +
    0.015
    *
    np.sin(
        np.linspace(
            0,
            8 * np.pi,
            TOTAL_SAMPLES
        )
    )
)


actual_values = (
    normal_values
    *
    rainfall_ratio
    *
    variation
)


X = normal_values.reshape(
    -1,
    1
)


y = actual_values.reshape(
    -1,
    1
)


print("\nTraining samples:", len(X))


# ======================================================================
# 11. TRAIN TEST SPLIT
# ======================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print(
    "Training Samples:",
    len(X_train)
)

print(
    "Testing Samples:",
    len(X_test)
)


# ======================================================================
# 12. SCALING
# ======================================================================

feature_scaler = StandardScaler()

target_scaler = StandardScaler()


X_train_scaled = feature_scaler.fit_transform(
    X_train
)


X_test_scaled = feature_scaler.transform(
    X_test
)


y_train_scaled = target_scaler.fit_transform(
    y_train
)


y_test_scaled = target_scaler.transform(
    y_test
)


# ======================================================================
# 13. BUILD ANN MODEL
# ======================================================================

model = Sequential(
    [
        Input(
            shape=(1,)
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            8,
            activation="relu"
        ),

        Dense(
            1,
            activation="linear"
        )
    ]
)


# ======================================================================
# 14. COMPILE MODEL
# ======================================================================

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=[
        "mae"
    ]
)


print("\n" + "=" * 75)
print("MODEL SUMMARY")
print("=" * 75)

model.summary()


# ======================================================================
# 15. EARLY STOPPING
# ======================================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=40,
    restore_best_weights=True,
    verbose=0
)


# ======================================================================
# 16. TRAIN MODEL
# ======================================================================

print("\n" + "=" * 75)
print("TRAINING MODEL")
print("=" * 75)


history = model.fit(
    X_train_scaled,
    y_train_scaled,

    validation_split=0.20,

    epochs=500,

    batch_size=16,

    callbacks=[
        early_stopping
    ],

    verbose=0
)


epochs_completed = len(
    history.history[
        "loss"
    ]
)


print(
    "\nTraining completed."
)

print(
    "Epochs:",
    epochs_completed
)


# ======================================================================
# 17. MAKE TEST PREDICTIONS
# ======================================================================

test_prediction_scaled = model.predict(
    X_test_scaled,
    verbose=0
)


test_predictions = target_scaler.inverse_transform(
    test_prediction_scaled
)


actual_test_values = y_test.flatten()

predicted_test_values = test_predictions.flatten()


# ======================================================================
# 18. MODEL METRICS
# ======================================================================

mae = mean_absolute_error(
    actual_test_values,
    predicted_test_values
)


mse = mean_squared_error(
    actual_test_values,
    predicted_test_values
)


rmse = np.sqrt(
    mse
)


r2 = r2_score(
    actual_test_values,
    predicted_test_values
)


safe_actual = np.where(
    actual_test_values == 0,
    1e-8,
    actual_test_values
)


mape = np.mean(
    np.abs(
        (
            actual_test_values
            -
            predicted_test_values
        )
        /
        safe_actual
    )
) * 100


regression_accuracy = max(
    0,
    100 - mape
)


print("\n" + "=" * 75)
print("MODEL PERFORMANCE")
print("=" * 75)


print(
    f"\nMAE      : {mae:.4f}"
)

print(
    f"MSE      : {mse:.4f}"
)

print(
    f"RMSE     : {rmse:.4f}"
)

print(
    f"R2 Score : {r2:.4f}"
)

print(
    f"MAPE     : {mape:.4f}%"
)

print(
    f"Regression Accuracy: {regression_accuracy:.4f}%"
)


# ======================================================================
# 19. SAVE H5 MODEL
# ======================================================================

model.save(
    MODEL_PATH
)


print(
    "\nModel saved:",
    MODEL_PATH
)


# ======================================================================
# 20. SAVE PKL PREPROCESSOR
# ======================================================================

preprocessor_data = {

    "feature_scaler":
        feature_scaler,

    "target_scaler":
        target_scaler,

    "normal_rainfall":
        normal_rainfall,

    "actual_rainfall":
        actual_rainfall,

    "rainfall_ratio":
        rainfall_ratio,

    "description_column":
        description_column,

    "rainfall_column":
        rainfall_column
}


joblib.dump(
    preprocessor_data,
    PKL_PATH
)


# ======================================================================
# 21. ACCURACY GRAPH
# ======================================================================
#
# Since this is regression, classification accuracy is not applicable.
# We plot regression accuracy derived from MAPE:
#
# Regression Accuracy = 100 - MAPE
#
# ======================================================================

print("\nGenerating accuracy graph...")


epochs = range(
    1,
    epochs_completed + 1
)


train_loss = np.array(
    history.history[
        "loss"
    ]
)


val_loss = np.array(
    history.history[
        "val_loss"
    ]
)


# Convert normalized MSE to an easy-to-visualize score.
train_accuracy_curve = (
    100
    /
    (
        1
        +
        train_loss
    )
)


val_accuracy_curve = (
    100
    /
    (
        1
        +
        val_loss
    )
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(
    epochs,
    train_accuracy_curve,
    linewidth=2,
    label="Training Score"
)


plt.plot(
    epochs,
    val_accuracy_curve,
    linewidth=2,
    label="Validation Score"
)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Regression Learning Score (%)"
)

plt.title(
    "Rainfall Prediction Model - Accuracy Graph"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    ACCURACY_GRAPH_PATH
)


# ======================================================================
# 22. HEATMAP
# ======================================================================
#
# Correlation heatmap using generated rainfall features.
# ======================================================================

print("\nGenerating heatmap...")


heatmap_df = pd.DataFrame(
    {
        "Normal Rainfall":
            normal_values,

        "Actual Rainfall":
            actual_values,

        "Rainfall Difference":
            actual_values
            -
            normal_values,

        "Rainfall Ratio":
            actual_values
            /
            normal_values,

        "Deviation (%)":
            (
                (
                    actual_values
                    -
                    normal_values
                )
                /
                normal_values
            )
            * 100
    }
)


correlation_matrix = heatmap_df.corr()


plt.figure(
    figsize=(10, 8)
)


plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    interpolation="nearest",
    vmin=-1,
    vmax=1
)


plt.colorbar(
    label="Correlation"
)


labels = correlation_matrix.columns.tolist()


plt.xticks(
    range(len(labels)),
    labels,
    rotation=45,
    ha="right"
)


plt.yticks(
    range(len(labels)),
    labels
)


for i in range(
    len(labels)
):

    for j in range(
        len(labels)
    ):

        value = correlation_matrix.iloc[
            i,
            j
        ]

        plt.text(
            j,
            i,
            f"{value:.2f}",
            ha="center",
            va="center"
        )


plt.title(
    "Rainfall Feature Correlation Heatmap"
)

plt.tight_layout()


plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    HEATMAP_PATH
)


# ======================================================================
# 23. COMPARISON GRAPH
# ======================================================================
#
# Compare important regression evaluation metrics.
# R2 is converted to percentage for visualization.
# ======================================================================

print("\nGenerating comparison graph...")


comparison_labels = [
    "Accuracy",
    "R2 Score",
    "MAPE"
]


comparison_values = [
    regression_accuracy,
    max(
        0,
        r2 * 100
    ),
    mape
]


plt.figure(
    figsize=(9, 6)
)


bars = plt.bar(
    comparison_labels,
    comparison_values
)


plt.ylabel(
    "Score (%)"
)

plt.title(
    "Rainfall Prediction Model Performance Comparison"
)


for bar, value in zip(
    bars,
    comparison_values
):

    plt.text(
        bar.get_x()
        +
        bar.get_width() / 2,

        bar.get_height(),

        f"{value:.2f}",

        ha="center",
        va="bottom",
        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)


# ======================================================================
# 24. CREATE RESULT CSV
# ======================================================================
#
# result.csv contains the test-set actual and predicted rainfall values.
# ======================================================================

result_df = pd.DataFrame(
    {
        "Normal_Rainfall_mm":
            X_test.flatten(),

        "Actual_Rainfall_mm":
            actual_test_values,

        "Predicted_Rainfall_mm":
            predicted_test_values
    }
)


result_df[
    "Absolute_Error_mm"
] = np.abs(
    result_df[
        "Actual_Rainfall_mm"
    ]
    -
    result_df[
        "Predicted_Rainfall_mm"
    ]
)


result_df[
    "Percentage_Error"
] = (
    result_df[
        "Absolute_Error_mm"
    ]
    /
    result_df[
        "Actual_Rainfall_mm"
    ]
) * 100


# Sort by normal rainfall for clean graph
result_df = result_df.sort_values(
    by="Normal_Rainfall_mm"
).reset_index(
    drop=True
)


result_df.insert(
    0,
    "Sample_ID",
    np.arange(
        1,
        len(result_df) + 1
    )
)


result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)


print(
    "\nResult CSV saved:"
)

print(
    RESULT_CSV_PATH
)


# ======================================================================
# 25. RESULT GRAPH
# ======================================================================

print("\nGenerating result graph...")


plt.figure(
    figsize=(12, 6)
)


plt.plot(
    result_df[
        "Sample_ID"
    ],

    result_df[
        "Actual_Rainfall_mm"
    ],

    linewidth=2,

    label="Actual Rainfall"
)


plt.plot(
    result_df[
        "Sample_ID"
    ],

    result_df[
        "Predicted_Rainfall_mm"
    ],

    linestyle="--",
    linewidth=2,

    label="Predicted Rainfall"
)


plt.xlabel(
    "Test Sample"
)

plt.ylabel(
    "Rainfall (mm)"
)

plt.title(
    "Actual vs Predicted Rainfall Results"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    RESULT_GRAPH_PATH
)


# ======================================================================
# 26. FUTURE / RANGE PREDICTIONS
# ======================================================================
#
# Generate a rainfall prediction range around the original normal
# rainfall value.
# ======================================================================

prediction_normal_values = np.linspace(
    normal_rainfall * 0.70,
    normal_rainfall * 1.30,
    50
)


prediction_input = (
    prediction_normal_values
    .reshape(
        -1,
        1
    )
)


prediction_input_scaled = (
    feature_scaler.transform(
        prediction_input
    )
)


prediction_scaled = model.predict(
    prediction_input_scaled,
    verbose=0
)


prediction_values = (
    target_scaler.inverse_transform(
        prediction_scaled
    )
    .flatten()
)


# ======================================================================
# 27. CREATE PREDICTION CSV
# ======================================================================

prediction_df = pd.DataFrame(
    {
        "Prediction_ID":
            np.arange(
                1,
                len(
                    prediction_normal_values
                )
                + 1
            ),

        "Normal_Rainfall_mm":
            prediction_normal_values,

        "Predicted_Actual_Rainfall_mm":
            prediction_values
    }
)


prediction_df[
    "Predicted_Difference_mm"
] = (
    prediction_df[
        "Predicted_Actual_Rainfall_mm"
    ]
    -
    prediction_df[
        "Normal_Rainfall_mm"
    ]
)


prediction_df[
    "Predicted_Deviation_Percentage"
] = (
    prediction_df[
        "Predicted_Difference_mm"
    ]
    /
    prediction_df[
        "Normal_Rainfall_mm"
    ]
) * 100


prediction_df[
    "Rainfall_Category"
] = np.where(
    prediction_df[
        "Predicted_Deviation_Percentage"
    ]
    > 0,

    "Above Normal",

    np.where(
        prediction_df[
            "Predicted_Deviation_Percentage"
        ]
        < 0,

        "Below Normal",

        "Normal"
    )
)


prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)


print(
    "\nPrediction CSV saved:"
)

print(
    PREDICTION_CSV_PATH
)


# ======================================================================
# 28. PREDICTION GRAPH
# ======================================================================

print("\nGenerating prediction graph...")


plt.figure(
    figsize=(11, 6)
)


plt.plot(
    prediction_df[
        "Normal_Rainfall_mm"
    ],

    prediction_df[
        "Predicted_Actual_Rainfall_mm"
    ],

    marker="o",
    markersize=4,
    linewidth=2,

    label="Predicted Actual Rainfall"
)


plt.plot(
    prediction_df[
        "Normal_Rainfall_mm"
    ],

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    linestyle="--",
    linewidth=2,

    label="Normal Rainfall Reference"
)


# Original observed point
plt.scatter(
    [
        normal_rainfall
    ],

    [
        actual_rainfall
    ],

    s=100,

    label="Original Observed Rainfall",

    zorder=5
)


plt.xlabel(
    "Normal Rainfall (mm)"
)

plt.ylabel(
    "Predicted Actual Rainfall (mm)"
)

plt.title(
    "Rainfall Prediction Graph"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)


# ======================================================================
# 29. ORIGINAL NORMAL RAINFALL PREDICTION
# ======================================================================

original_input = np.array(
    [
        [
            normal_rainfall
        ]
    ]
)


original_scaled = feature_scaler.transform(
    original_input
)


original_prediction_scaled = model.predict(
    original_scaled,
    verbose=0
)


original_prediction = float(
    target_scaler.inverse_transform(
        original_prediction_scaled
    )[0][0]
)


print("\n" + "=" * 75)
print("ORIGINAL DATA PREDICTION")
print("=" * 75)


print(
    f"\nNormal Rainfall   : {normal_rainfall:.2f} mm"
)

print(
    f"Actual Rainfall   : {actual_rainfall:.2f} mm"
)

print(
    f"Predicted Rainfall: {original_prediction:.2f} mm"
)


# ======================================================================
# 30. SAVE YAML CONFIGURATION
# ======================================================================

yaml_data = {

    "project": {

        "name":
            "AI-Based Rainfall Pattern Analysis and Prediction",

        "version":
            "1.0"
    },

    "dataset": {

        "path":
            DATASET_PATH,

        "rows":
            int(
                len(df)
            ),

        "columns":
            int(
                len(df.columns)
            )
    },

    "rainfall": {

        "normal_mm":
            float(
                normal_rainfall
            ),

        "actual_mm":
            float(
                actual_rainfall
            ),

        "difference_mm":
            float(
                difference
            ),

        "deviation_percentage":
            float(
                percentage_deviation
            ),

        "pattern":
            rainfall_pattern
    },

    "model": {

        "type":
            "Artificial Neural Network Regression",

        "framework":
            "TensorFlow Keras",

        "training_samples":
            int(
                len(X_train)
            ),

        "testing_samples":
            int(
                len(X_test)
            ),

        "epochs":
            int(
                epochs_completed
            ),

        "batch_size":
            16
    },

    "performance": {

        "mae":
            float(
                mae
            ),

        "mse":
            float(
                mse
            ),

        "rmse":
            float(
                rmse
            ),

        "r2_score":
            float(
                r2
            ),

        "mape_percentage":
            float(
                mape
            ),

        "regression_accuracy_percentage":
            float(
                regression_accuracy
            )
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_data,
        file,
        sort_keys=False
    )


# ======================================================================
# 31. SAVE JSON METADATA
# ======================================================================

json_data = {

    "project_name":
        "AI-Based Rainfall Pattern Analysis and Prediction",

    "normal_rainfall_mm":
        round(
            normal_rainfall,
            4
        ),

    "actual_rainfall_mm":
        round(
            actual_rainfall,
            4
        ),

    "predicted_original_rainfall_mm":
        round(
            original_prediction,
            4
        ),

    "rainfall_pattern":
        rainfall_pattern,

    "model_metrics": {

        "MAE":
            round(
                mae,
                6
            ),

        "MSE":
            round(
                mse,
                6
            ),

        "RMSE":
            round(
                rmse,
                6
            ),

        "R2":
            round(
                r2,
                6
            ),

        "MAPE":
            round(
                mape,
                6
            ),

        "Regression_Accuracy":
            round(
                regression_accuracy,
                6
            )
    },

    "generated_files": [

        "rainfall_prediction_model.h5",

        "rainfall_preprocessor.pkl",

        "rainfall_config.yaml",

        "rainfall_metadata.json",

        "accuracy_graph.png",

        "heatmap.png",

        "comparison_graph.png",

        "result.csv",

        "result_graph.png",

        "prediction.csv",

        "prediction_graph.png"
    ],

    "note":
        (
            "The supplied dataset contains only aggregate Normal and "
            "Actual rainfall observations. Synthetic local samples were "
            "used to demonstrate the machine-learning workflow. "
            "A larger historical dataset is required for reliable "
            "real-world rainfall forecasting."
        )
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_data,
        file,
        indent=4
    )


# ======================================================================
# 32. DISPLAY RESULT CSV
# ======================================================================

print("\n" + "=" * 75)
print("RESULT CSV PREVIEW")
print("=" * 75)

print(
    result_df.head(
        10
    ).to_string(
        index=False
    )
)


# ======================================================================
# 33. DISPLAY PREDICTION CSV
# ======================================================================

print("\n" + "=" * 75)
print("PREDICTION CSV PREVIEW")
print("=" * 75)

print(
    prediction_df.head(
        10
    ).to_string(
        index=False
    )
)


# ======================================================================
# 34. VERIFY ALL FILES
# ======================================================================

generated_files = [

    MODEL_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH,

    ACCURACY_GRAPH_PATH,
    HEATMAP_PATH,
    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,
    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,
    PREDICTION_GRAPH_PATH
]


print("\n" + "=" * 75)
print("GENERATED FILES")
print("=" * 75)


for file_path in generated_files:

    if os.path.exists(
        file_path
    ):

        size_kb = (
            os.path.getsize(
                file_path
            )
            /
            1024
        )

        print(
            f"[OK] "
            f"{os.path.basename(file_path)} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            f"[ERROR] "
            f"{os.path.basename(file_path)}"
        )


# ======================================================================
# 35. FINAL SUMMARY
# ======================================================================

print("\n" + "=" * 75)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 75)


print(
    f"""
Project:
AI-Based Rainfall Pattern Analysis and Prediction

Normal Rainfall:
{normal_rainfall:.2f} mm

Actual Rainfall:
{actual_rainfall:.2f} mm

Observed Pattern:
{rainfall_pattern}

Model Performance
-----------------
MAE:
{mae:.4f}

RMSE:
{rmse:.4f}

R2 Score:
{r2:.4f}

MAPE:
{mape:.4f}%

Regression Accuracy:
{regression_accuracy:.4f}%

Generated Files
---------------
1. rainfall_prediction_model.h5
2. rainfall_preprocessor.pkl
3. rainfall_config.yaml
4. rainfall_metadata.json
5. accuracy_graph.png
6. heatmap.png
7. comparison_graph.png
8. result.csv
9. result_graph.png
10. prediction.csv
11. prediction_graph.png

Output Directory:
{PROJECT_DIR}
"""
)


print("=" * 75)
```
