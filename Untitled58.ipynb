```python
import os
import json
import random
import warnings
import gc

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

import joblib
import yaml

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Input,
    Dense
)

from tensorflow.keras.callbacks import (
    EarlyStopping
)


# ======================================================================
# 2. SETTINGS
# ======================================================================

warnings.filterwarnings(
    "ignore"
)

SEED = 42

np.random.seed(
    SEED
)

random.seed(
    SEED
)

tf.random.set_seed(
    SEED
)


# ======================================================================
# 3. PROJECT PATHS
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Rainfall Pattern Analysis and Prediction"
)


DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_june_2019_to_may_2020.csv"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ======================================================================
# 4. PSO OUTPUT PATHS
# ======================================================================

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "pso_rainfall_prediction_model.h5"
)


PKL_PATH = os.path.join(
    PROJECT_DIR,
    "pso_rainfall_preprocessor.pkl"
)


YAML_PATH = os.path.join(
    PROJECT_DIR,
    "pso_rainfall_config.yaml"
)


JSON_PATH = os.path.join(
    PROJECT_DIR,
    "pso_rainfall_metadata.json"
)


ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_accuracy_graph.png"
)


HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "pso_heatmap.png"
)


COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_comparison_graph.png"
)


RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result.csv"
)


RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result_graph.png"
)


PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction.csv"
)


PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction_graph.png"
)


PSO_HISTORY_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_optimization_history.csv"
)


PSO_OPTIMIZATION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_optimization_graph.png"
)


# ======================================================================
# 5. DISPLAY PROJECT INFORMATION
# ======================================================================

print(
    "=" * 78
)

print(
    "AI-BASED RAINFALL PATTERN ANALYSIS AND PREDICTION"
)

print(
    "PARTICLE SWARM OPTIMIZATION (PSO)"
)

print(
    "=" * 78
)


print(
    "\nProject Directory:"
)

print(
    PROJECT_DIR
)


print(
    "\nDataset:"
)

print(
    DATASET_PATH
)


# ======================================================================
# 6. CHECK DATASET
# ======================================================================

if not os.path.exists(
    DATASET_PATH
):

    raise FileNotFoundError(
        f"\nDataset not found:\n"
        f"{DATASET_PATH}"
    )


print(
    "\nDataset found successfully."
)


# ======================================================================
# 7. LOAD DATASET
# ======================================================================

df = pd.read_csv(
    DATASET_PATH
)


df.columns = (
    df.columns
    .str.strip()
)


print(
    "\n" + "=" * 78
)

print(
    "DATASET INFORMATION"
)

print(
    "=" * 78
)


print(
    "\nDataset Shape:",
    df.shape
)


print(
    "\nColumns:"
)


for column in df.columns:

    print(
        "-",
        column
    )


print(
    "\nDataset:"
)


print(
    df.to_string(
        index=False
    )
)


# ======================================================================
# 8. IDENTIFY COLUMNS
# ======================================================================

description_column = None

rainfall_column = None


for column in df.columns:

    lower_column = (
        column.lower()
    )

    if "description" in lower_column:

        description_column = column

    if "rainfall" in lower_column:

        rainfall_column = column


# ----------------------------------------------------------------------
# Description fallback
# ----------------------------------------------------------------------

if description_column is None:

    description_column = (
        df.columns[0]
    )


# ----------------------------------------------------------------------
# Rainfall fallback
# ----------------------------------------------------------------------

if rainfall_column is None:

    numeric_candidates = []


    for column in df.columns:

        converted = pd.to_numeric(
            df[column],
            errors="coerce"
        )


        if (
            converted
            .notna()
            .sum()
            > 0
        ):

            numeric_candidates.append(
                column
            )


    if len(
        numeric_candidates
    ) > 0:

        rainfall_column = (
            numeric_candidates[0]
        )


if rainfall_column is None:

    raise ValueError(
        "Rainfall column could not be identified."
    )


print(
    "\nDescription Column:",
    description_column
)


print(
    "Rainfall Column:",
    rainfall_column
)


# ======================================================================
# 9. CLEAN DATA
# ======================================================================

df[
    description_column
] = (
    df[
        description_column
    ]
    .astype(str)
    .str.strip()
)


df[
    rainfall_column
] = pd.to_numeric(
    df[
        rainfall_column
    ],
    errors="coerce"
)


df = (
    df
    .dropna(
        subset=[
            rainfall_column
        ]
    )
    .reset_index(
        drop=True
    )
)


# ======================================================================
# 10. EXTRACT NORMAL RAINFALL
# ======================================================================

normal_rows = df[
    df[
        description_column
    ]
    .str.lower()
    .str.contains(
        "normal",
        na=False
    )
]


if len(
    normal_rows
) == 0:

    raise ValueError(
        "Normal Rainfall row was not found."
    )


normal_rainfall = float(
    normal_rows[
        rainfall_column
    ].iloc[0]
)


# ======================================================================
# 11. EXTRACT ACTUAL RAINFALL
# ======================================================================

actual_rows = df[
    df[
        description_column
    ]
    .str.lower()
    .str.contains(
        "actual",
        na=False
    )
]


if len(
    actual_rows
) == 0:

    raise ValueError(
        "Actual Rainfall row was not found."
    )


actual_rainfall = float(
    actual_rows[
        rainfall_column
    ].iloc[0]
)


# ======================================================================
# 12. BASIC RAINFALL ANALYSIS
# ======================================================================

difference = (
    actual_rainfall
    -
    normal_rainfall
)


percentage_deviation = (
    difference
    /
    normal_rainfall
) * 100


rainfall_ratio = (
    actual_rainfall
    /
    normal_rainfall
)


if percentage_deviation > 0:

    rainfall_pattern = (
        "Above Normal"
    )

elif percentage_deviation < 0:

    rainfall_pattern = (
        "Below Normal"
    )

else:

    rainfall_pattern = (
        "Normal"
    )


print(
    "\n" + "=" * 78
)

print(
    "RAINFALL ANALYSIS"
)

print(
    "=" * 78
)


print(
    f"\nNormal Rainfall      : "
    f"{normal_rainfall:.2f} mm"
)


print(
    f"Actual Rainfall      : "
    f"{actual_rainfall:.2f} mm"
)


print(
    f"Difference           : "
    f"{difference:.2f} mm"
)


print(
    f"Percentage Deviation : "
    f"{percentage_deviation:.4f}%"
)


print(
    f"Rainfall Ratio       : "
    f"{rainfall_ratio:.6f}"
)


print(
    f"Rainfall Pattern     : "
    f"{rainfall_pattern}"
)


# ======================================================================
# 13. CREATE TRAINING DATA
# ======================================================================
#
# IMPORTANT:
#
# The supplied CSV contains an aggregate Normal/Actual rainfall pair.
# This is not enough to independently train and evaluate an ANN.
#
# Therefore, a local synthetic demonstration dataset is generated around
# the observed Normal-to-Actual rainfall relationship.
#
# PSO optimizes the ANN against this demonstration dataset.
#
# ======================================================================

TOTAL_SAMPLES = 500


training_normal_values = np.linspace(

    normal_rainfall * 0.60,

    normal_rainfall * 1.40,

    TOTAL_SAMPLES
)


variation = (

    1.0

    +

    0.015

    *

    np.sin(

        np.linspace(
            0,
            8 * np.pi,
            TOTAL_SAMPLES
        )
    )
)


training_actual_values = (

    training_normal_values

    *

    rainfall_ratio

    *

    variation
)


X = (
    training_normal_values
    .reshape(
        -1,
        1
    )
)


y = (
    training_actual_values
    .reshape(
        -1,
        1
    )
)


print(
    "\nTotal Generated Samples:",
    len(X)
)


# ======================================================================
# 14. TRAIN TEST SPLIT
# ======================================================================

(
    X_train,
    X_test,
    y_train,
    y_test

) = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=SEED
)


print(
    "Training Samples:",
    len(X_train)
)


print(
    "Testing Samples:",
    len(X_test)
)


# ======================================================================
# 15. CREATE SCALERS
# ======================================================================

feature_scaler = (
    StandardScaler()
)


target_scaler = (
    StandardScaler()
)


# ======================================================================
# 16. SCALE DATA
# ======================================================================

X_train_scaled = (
    feature_scaler
    .fit_transform(
        X_train
    )
)


X_test_scaled = (
    feature_scaler
    .transform(
        X_test
    )
)


y_train_scaled = (
    target_scaler
    .fit_transform(
        y_train
    )
)


y_test_scaled = (
    target_scaler
    .transform(
        y_test
    )
)


# ======================================================================
# 17. CREATE PSO TRAINING / VALIDATION DATA
# ======================================================================

(
    X_pso_train,
    X_pso_validation,
    y_pso_train,
    y_pso_validation

) = train_test_split(

    X_train_scaled,
    y_train_scaled,

    test_size=0.20,

    random_state=SEED
)


# ======================================================================
# 18. ANN CREATION FUNCTION
# ======================================================================

def create_ann_model(
    neurons_1,
    neurons_2,
    learning_rate
):

    """
    Create an ANN using hyperparameters supplied by PSO.
    """

    model = Sequential(
        [

            Input(
                shape=(1,)
            ),

            Dense(
                int(
                    neurons_1
                ),
                activation="relu"
            ),

            Dense(
                int(
                    neurons_2
                ),
                activation="relu"
            ),

            Dense(
                8,
                activation="relu"
            ),

            Dense(
                1,
                activation="linear"
            )

        ]
    )


    # --------------------------------------------------------------
    # H5 COMPATIBILITY FIX
    # --------------------------------------------------------------
    #
    # Explicit Keras loss/metric objects are used rather than:
    #
    # loss="mse"
    #
    # --------------------------------------------------------------

    model.compile(

        optimizer=tf.keras.optimizers.Adam(
            learning_rate=float(
                learning_rate
            )
        ),

        loss=tf.keras.losses.MeanSquaredError(),

        metrics=[
            tf.keras.metrics.MeanAbsoluteError(
                name="mae"
            )
        ]
    )


    return model


# ======================================================================
# 19. PSO SEARCH SPACE
# ======================================================================
#
# Particle dimensions:
#
# position[0] = neurons in first hidden layer
# position[1] = neurons in second hidden layer
# position[2] = learning rate
# position[3] = batch-size index
#
# ======================================================================

LOWER_BOUNDS = np.array(
    [
        16.0,
        8.0,
        0.0003,
        0.0
    ]
)


UPPER_BOUNDS = np.array(
    [
        96.0,
        64.0,
        0.005,
        2.0
    ]
)


BATCH_OPTIONS = [
    8,
    16,
    32
]


# ======================================================================
# 20. DECODE PARTICLE
# ======================================================================

def decode_particle(
    position
):

    """
    Convert continuous PSO position into valid ANN hyperparameters.
    """

    neurons_1 = int(
        np.clip(
            round(
                position[0]
            ),
            16,
            96
        )
    )


    neurons_2 = int(
        np.clip(
            round(
                position[1]
            ),
            8,
            64
        )
    )


    learning_rate = float(
        np.clip(
            position[2],
            0.0003,
            0.005
        )
    )


    batch_index = int(
        np.clip(
            round(
                position[3]
            ),
            0,
            len(
                BATCH_OPTIONS
            ) - 1
        )
    )


    batch_size = (
        BATCH_OPTIONS[
            batch_index
        ]
    )


    return (
        neurons_1,
        neurons_2,
        learning_rate,
        batch_size
    )


# ======================================================================
# 21. PSO FITNESS FUNCTION
# ======================================================================

def evaluate_particle(
    position
):

    """
    Evaluate one PSO particle.

    The objective is validation MSE.

    Lower fitness is better.
    """

    (
        neurons_1,
        neurons_2,
        learning_rate,
        batch_size

    ) = decode_particle(
        position
    )


    tf.keras.backend.clear_session()

    gc.collect()


    model = create_ann_model(

        neurons_1,

        neurons_2,

        learning_rate
    )


    early_stop = EarlyStopping(

        monitor="val_loss",

        patience=10,

        restore_best_weights=True,

        verbose=0
    )


    history = model.fit(

        X_pso_train,

        y_pso_train,

        validation_data=(
            X_pso_validation,
            y_pso_validation
        ),

        epochs=80,

        batch_size=batch_size,

        callbacks=[
            early_stop
        ],

        verbose=0
    )


    validation_loss = float(
        min(
            history.history[
                "val_loss"
            ]
        )
    )


    del model

    tf.keras.backend.clear_session()

    gc.collect()


    return validation_loss


# ======================================================================
# 22. PSO PARAMETERS
# ======================================================================
#
# Moderate values are used so that the optimization can run on a
# standard laptop without becoming unnecessarily expensive.
#
# Increase these if you want a larger search.
#
# ======================================================================

SWARM_SIZE = 6

PSO_ITERATIONS = 5


# Inertia weight
W = 0.70


# Cognitive coefficient
C1 = 1.50


# Social coefficient
C2 = 1.50


# ======================================================================
# 23. INITIALIZE SWARM
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "STARTING PARTICLE SWARM OPTIMIZATION"
)

print(
    "=" * 78
)


dimensions = len(
    LOWER_BOUNDS
)


positions = np.random.uniform(

    LOWERS := LOWER_BOUNDS,

    UPPERS := UPPER_BOUNDS,

    size=(
        SWARM_SIZE,
        dimensions
    )
)


velocity_ranges = (
    UPPER_BOUNDS
    -
    LOWER_BOUNDS
)


velocities = np.random.uniform(

    -0.10 * velocity_ranges,

    0.10 * velocity_ranges,

    size=(
        SWARM_SIZE,
        dimensions
    )
)


# ======================================================================
# 24. PERSONAL BEST INITIALIZATION
# ======================================================================

personal_best_positions = (
    positions.copy()
)


personal_best_scores = np.full(
    SWARM_SIZE,
    np.inf
)


global_best_position = None

global_best_score = np.inf


# ======================================================================
# 25. PSO HISTORY
# ======================================================================

pso_history = []


# ======================================================================
# 26. EVALUATE INITIAL SWARM
# ======================================================================

print(
    "\nEvaluating Initial Swarm..."
)


for particle_index in range(
    SWARM_SIZE
):

    score = evaluate_particle(
        positions[
            particle_index
        ]
    )


    personal_best_scores[
        particle_index
    ] = score


    personal_best_positions[
        particle_index
    ] = positions[
        particle_index
    ].copy()


    if score < global_best_score:

        global_best_score = (
            score
        )

        global_best_position = (
            positions[
                particle_index
            ]
            .copy()
        )


    (
        n1,
        n2,
        lr,
        bs

    ) = decode_particle(
        positions[
            particle_index
        ]
    )


    print(
        f"\nParticle "
        f"{particle_index + 1}/{SWARM_SIZE}"
    )


    print(
        f"Neurons 1      : {n1}"
    )


    print(
        f"Neurons 2      : {n2}"
    )


    print(
        f"Learning Rate  : {lr:.8f}"
    )


    print(
        f"Batch Size     : {bs}"
    )


    print(
        f"Validation Loss: {score:.8f}"
    )


# ======================================================================
# 27. PSO OPTIMIZATION LOOP
# ======================================================================

for iteration in range(
    PSO_ITERATIONS
):

    print(
        "\n" + "=" * 78
    )

    print(
        f"PSO ITERATION "
        f"{iteration + 1}/{PSO_ITERATIONS}"
    )

    print(
        "=" * 78
    )


    iteration_scores = []


    for particle_index in range(
        SWARM_SIZE
    ):

        # ----------------------------------------------------------
        # Random PSO factors
        # ----------------------------------------------------------

        r1 = np.random.random(
            dimensions
        )


        r2 = np.random.random(
            dimensions
        )


        # ----------------------------------------------------------
        # Cognitive component
        # ----------------------------------------------------------

        cognitive_component = (

            C1

            *

            r1

            *

            (
                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]
            )
        )


        # ----------------------------------------------------------
        # Social component
        # ----------------------------------------------------------

        social_component = (

            C2

            *

            r2

            *

            (
                global_best_position

                -

                positions[
                    particle_index
                ]
            )
        )


        # ----------------------------------------------------------
        # Velocity update
        # ----------------------------------------------------------

        velocities[
            particle_index
        ] = (

            W
            *
            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component
        )


        # ----------------------------------------------------------
        # Velocity clipping
        # ----------------------------------------------------------

        max_velocity = (
            0.25
            *
            velocity_ranges
        )


        velocities[
            particle_index
        ] = np.clip(

            velocities[
                particle_index
            ],

            -max_velocity,

            max_velocity
        )


        # ----------------------------------------------------------
        # Position update
        # ----------------------------------------------------------

        positions[
            particle_index
        ] = (

            positions[
                particle_index
            ]

            +

            velocities[
                particle_index
            ]
        )


        # ----------------------------------------------------------
        # Keep position within search bounds
        # ----------------------------------------------------------

        positions[
            particle_index
        ] = np.clip(

            positions[
                particle_index
            ],

            LOWER_BOUNDS,

            UPPER_BOUNDS
        )


        # ----------------------------------------------------------
        # Evaluate particle
        # ----------------------------------------------------------

        score = evaluate_particle(

            positions[
                particle_index
            ]
        )


        iteration_scores.append(
            score
        )


        # ----------------------------------------------------------
        # Update personal best
        # ----------------------------------------------------------

        if (
            score
            <
            personal_best_scores[
                particle_index
            ]
        ):

            personal_best_scores[
                particle_index
            ] = score


            personal_best_positions[
                particle_index
            ] = (

                positions[
                    particle_index
                ]
                .copy()
            )


        # ----------------------------------------------------------
        # Update global best
        # ----------------------------------------------------------

        if score < global_best_score:

            global_best_score = (
                score
            )


            global_best_position = (

                positions[
                    particle_index
                ]
                .copy()
            )


        (
            current_n1,
            current_n2,
            current_lr,
            current_bs

        ) = decode_particle(

            positions[
                particle_index
            ]
        )


        print(
            f"\nParticle "
            f"{particle_index + 1}"
        )


        print(
            f"Neurons 1      : "
            f"{current_n1}"
        )


        print(
            f"Neurons 2      : "
            f"{current_n2}"
        )


        print(
            f"Learning Rate  : "
            f"{current_lr:.8f}"
        )


        print(
            f"Batch Size     : "
            f"{current_bs}"
        )


        print(
            f"Validation Loss: "
            f"{score:.8f}"
        )


    # --------------------------------------------------------------
    # Decode global best
    # --------------------------------------------------------------

    (
        best_n1,
        best_n2,
        best_lr,
        best_bs

    ) = decode_particle(
        global_best_position
    )


    average_score = float(
        np.mean(
            iteration_scores
        )
    )


    pso_history.append(
        {

            "Iteration":
                iteration + 1,

            "Global_Best_Loss":
                global_best_score,

            "Average_Loss":
                average_score,

            "Best_Neurons_1":
                best_n1,

            "Best_Neurons_2":
                best_n2,

            "Best_Learning_Rate":
                best_lr,

            "Best_Batch_Size":
                best_bs

        }
    )


    print(
        "\nGLOBAL BEST"
    )


    print(
        f"Validation Loss: "
        f"{global_best_score:.8f}"
    )


    print(
        f"Neurons 1      : "
        f"{best_n1}"
    )


    print(
        f"Neurons 2      : "
        f"{best_n2}"
    )


    print(
        f"Learning Rate  : "
        f"{best_lr:.8f}"
    )


    print(
        f"Batch Size     : "
        f"{best_bs}"
    )


# ======================================================================
# 28. FINAL BEST PSO PARAMETERS
# ======================================================================

(
    BEST_NEURONS_1,
    BEST_NEURONS_2,
    BEST_LEARNING_RATE,
    BEST_BATCH_SIZE

) = decode_particle(
    global_best_position
)


print(
    "\n" + "=" * 78
)

print(
    "PSO OPTIMIZATION COMPLETED"
)

print(
    "=" * 78
)


print(
    f"\nBest First Hidden Layer : "
    f"{BEST_NEURONS_1}"
)


print(
    f"Best Second Hidden Layer: "
    f"{BEST_NEURONS_2}"
)


print(
    f"Best Learning Rate      : "
    f"{BEST_LEARNING_RATE:.8f}"
)


print(
    f"Best Batch Size         : "
    f"{BEST_BATCH_SIZE}"
)


print(
    f"Best Validation Loss    : "
    f"{global_best_score:.8f}"
)


# ======================================================================
# 29. SAVE PSO HISTORY CSV
# ======================================================================

pso_history_df = pd.DataFrame(
    pso_history
)


pso_history_df.to_csv(
    PSO_HISTORY_CSV_PATH,
    index=False
)


print(
    "\nPSO Optimization History Saved:"
)

print(
    PSO_HISTORY_CSV_PATH
)


# ======================================================================
# 30. PSO OPTIMIZATION GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    pso_history_df[
        "Iteration"
    ],

    pso_history_df[
        "Global_Best_Loss"
    ],

    marker="o",

    linewidth=2,

    label="Global Best Loss"
)


plt.plot(

    pso_history_df[
        "Iteration"
    ],

    pso_history_df[
        "Average_Loss"
    ],

    marker="s",

    linewidth=2,

    label="Average Swarm Loss"
)


plt.xlabel(
    "PSO Iteration"
)


plt.ylabel(
    "Validation Loss"
)


plt.title(
    "PSO Optimization Progress"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    PSO_OPTIMIZATION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nPSO Optimization Graph Saved:"
)

print(
    PSO_OPTIMIZATION_GRAPH_PATH
)


# ======================================================================
# 31. CLEAR TEMPORARY MODELS
# ======================================================================

tf.keras.backend.clear_session()

gc.collect()


# ======================================================================
# 32. CREATE FINAL PSO-OPTIMIZED ANN
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "TRAINING FINAL PSO-OPTIMIZED ANN"
)

print(
    "=" * 78
)


final_model = create_ann_model(

    BEST_NEURONS_1,

    BEST_NEURONS_2,

    BEST_LEARNING_RATE
)


final_model.summary()


# ======================================================================
# 33. FINAL EARLY STOPPING
# ======================================================================

final_early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=40,

    restore_best_weights=True,

    verbose=0
)


# ======================================================================
# 34. TRAIN FINAL PSO MODEL
# ======================================================================

final_history = final_model.fit(

    X_train_scaled,

    y_train_scaled,

    validation_split=0.20,

    epochs=500,

    batch_size=BEST_BATCH_SIZE,

    callbacks=[
        final_early_stopping
    ],

    verbose=0
)


epochs_completed = len(
    final_history.history[
        "loss"
    ]
)


print(
    "\nFinal PSO model training completed."
)


print(
    "Epochs Completed:",
    epochs_completed
)


# ======================================================================
# 35. TEST PREDICTIONS
# ======================================================================

test_prediction_scaled = (
    final_model.predict(
        X_test_scaled,
        verbose=0
    )
)


test_predictions = (
    target_scaler
    .inverse_transform(
        test_prediction_scaled
    )
)


actual_test_values = (
    y_test.flatten()
)


predicted_test_values = (
    test_predictions.flatten()
)


# ======================================================================
# 36. PSO MODEL METRICS
# ======================================================================

mae = mean_absolute_error(

    actual_test_values,

    predicted_test_values
)


mse = mean_squared_error(

    actual_test_values,

    predicted_test_values
)


rmse = np.sqrt(
    mse
)


r2 = r2_score(

    actual_test_values,

    predicted_test_values
)


safe_actual = np.where(

    actual_test_values == 0,

    1e-8,

    actual_test_values
)


mape = (

    np.mean(

        np.abs(

            (
                actual_test_values
                -
                predicted_test_values
            )

            /

            safe_actual
        )
    )

    *

    100
)


regression_accuracy = max(

    0.0,

    100.0
    -
    mape
)


print(
    "\n" + "=" * 78
)

print(
    "PSO MODEL PERFORMANCE"
)

print(
    "=" * 78
)


print(
    f"\nMAE                 : "
    f"{mae:.6f}"
)


print(
    f"MSE                 : "
    f"{mse:.6f}"
)


print(
    f"RMSE                : "
    f"{rmse:.6f}"
)


print(
    f"R2 Score            : "
    f"{r2:.6f}"
)


print(
    f"MAPE                : "
    f"{mape:.6f}%"
)


print(
    f"Regression Accuracy : "
    f"{regression_accuracy:.6f}%"
)


# ======================================================================
# 37. ORIGINAL RAINFALL PREDICTION
# ======================================================================

original_input = np.array(
    [
        [
            normal_rainfall
        ]
    ],
    dtype=np.float32
)


original_scaled = (
    feature_scaler
    .transform(
        original_input
    )
)


original_prediction_scaled = (
    final_model.predict(
        original_scaled,
        verbose=0
    )
)


predicted_actual_rainfall = float(

    target_scaler
    .inverse_transform(
        original_prediction_scaled
    )[0][0]
)


prediction_error = abs(

    actual_rainfall

    -

    predicted_actual_rainfall
)


percentage_error = (

    prediction_error

    /

    actual_rainfall

) * 100


print(
    "\n" + "=" * 78
)

print(
    "PSO ORIGINAL RAINFALL PREDICTION"
)

print(
    "=" * 78
)


print(
    f"\nNormal Rainfall    : "
    f"{normal_rainfall:.2f} mm"
)


print(
    f"Actual Rainfall    : "
    f"{actual_rainfall:.2f} mm"
)


print(
    f"PSO Prediction     : "
    f"{predicted_actual_rainfall:.2f} mm"
)


print(
    f"Absolute Error     : "
    f"{prediction_error:.6f} mm"
)


print(
    f"Percentage Error   : "
    f"{percentage_error:.6f}%"
)


# ======================================================================
# 38. SAVE PSO MODEL
# ======================================================================

final_model.save(
    MODEL_PATH
)


print(
    "\nPSO H5 Model Saved:"
)

print(
    MODEL_PATH
)


# ======================================================================
# 39. SAVE PSO PREPROCESSOR
# ======================================================================

preprocessor_data = {

    "feature_scaler":
        feature_scaler,

    "target_scaler":
        target_scaler,

    "feature_name":
        "Normal Rainfall",

    "target_name":
        "Actual Rainfall",

    "description_column":
        description_column,

    "rainfall_column":
        rainfall_column,

    "normal_rainfall":
        normal_rainfall,

    "actual_rainfall":
        actual_rainfall,

    "rainfall_ratio":
        rainfall_ratio,

    "optimization":
        "Particle Swarm Optimization",

    "best_neurons_1":
        BEST_NEURONS_1,

    "best_neurons_2":
        BEST_NEURONS_2,

    "best_learning_rate":
        BEST_LEARNING_RATE,

    "best_batch_size":
        BEST_BATCH_SIZE,

    "best_validation_loss":
        global_best_score

}


joblib.dump(
    preprocessor_data,
    PKL_PATH
)


print(
    "\nPSO PKL Saved:"
)

print(
    PKL_PATH
)


# ======================================================================
# 40. PSO ACCURACY GRAPH
# ======================================================================
#
# Since this is regression, standard classification accuracy does not
# apply. A learning score derived from loss is displayed.
#
# ======================================================================

training_loss = np.array(
    final_history.history[
        "loss"
    ]
)


validation_loss = np.array(
    final_history.history[
        "val_loss"
    ]
)


training_score = (

    100.0

    /

    (
        1.0
        +
        training_loss
    )
)


validation_score = (

    100.0

    /

    (
        1.0
        +
        validation_loss
    )
)


epoch_numbers = np.arange(

    1,

    epochs_completed + 1
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    epoch_numbers,

    training_score,

    linewidth=2,

    label="PSO Training Score"
)


plt.plot(

    epoch_numbers,

    validation_score,

    linewidth=2,

    label="PSO Validation Score"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Learning Score (%)"
)


plt.title(
    "PSO-Optimized Rainfall Model Accuracy"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nPSO Accuracy Graph Saved:"
)

print(
    ACCURACY_GRAPH_PATH
)


# ======================================================================
# 41. CREATE CORRELATION DATA
# ======================================================================

correlation_df = pd.DataFrame(
    {

        "Normal Rainfall":
            training_normal_values,

        "Actual Rainfall":
            training_actual_values,

        "Difference":
            (
                training_actual_values
                -
                training_normal_values
            ),

        "Ratio":
            (
                training_actual_values
                /
                training_normal_values
            ),

        "Deviation":
            (
                (
                    training_actual_values
                    -
                    training_normal_values
                )

                /

                training_normal_values
            )
            * 100

    }
)


correlation_matrix = (
    correlation_df
    .corr()
)


# ======================================================================
# 42. PSO HEATMAP
# ======================================================================

plt.figure(
    figsize=(10, 8)
)


heatmap_image = plt.imshow(

    correlation_matrix,

    cmap="coolwarm",

    interpolation="nearest",

    vmin=-1,

    vmax=1
)


plt.colorbar(
    heatmap_image,
    label="Correlation"
)


heatmap_labels = (
    correlation_matrix
    .columns
    .tolist()
)


plt.xticks(

    range(
        len(
            heatmap_labels
        )
    ),

    heatmap_labels,

    rotation=45,

    ha="right"
)


plt.yticks(

    range(
        len(
            heatmap_labels
        )
    ),

    heatmap_labels
)


for i in range(
    len(
        heatmap_labels
    )
):

    for j in range(
        len(
            heatmap_labels
        )
    ):

        value = (
            correlation_matrix
            .iloc[
                i,
                j
            ]
        )


        plt.text(

            j,

            i,

            f"{value:.2f}",

            ha="center",

            va="center"
        )


plt.title(
    "PSO Rainfall Feature Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nPSO Heatmap Saved:"
)

print(
    HEATMAP_PATH
)


# ======================================================================
# 43. PSO COMPARISON GRAPH
# ======================================================================

comparison_labels = [

    "Regression Accuracy",

    "R2 Score",

    "MAPE"
]


comparison_values = [

    regression_accuracy,

    max(
        0.0,
        r2 * 100.0
    ),

    mape
]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(

    comparison_labels,

    comparison_values
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "PSO Rainfall Model Performance Comparison"
)


for bar, value in zip(
    bars,
    comparison_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        bar.get_height(),

        f"{value:.2f}",

        ha="center",

        va="bottom",

        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nPSO Comparison Graph Saved:"
)

print(
    COMPARISON_GRAPH_PATH
)


# ======================================================================
# 44. CREATE PSO RESULT CSV
# ======================================================================

result_df = pd.DataFrame(
    {

        "Normal_Rainfall_mm":
            X_test.flatten(),

        "Actual_Rainfall_mm":
            actual_test_values,

        "PSO_Predicted_Rainfall_mm":
            predicted_test_values

    }
)


result_df[
    "PSO_Absolute_Error_mm"
] = np.abs(

    result_df[
        "Actual_Rainfall_mm"
    ]

    -

    result_df[
        "PSO_Predicted_Rainfall_mm"
    ]
)


result_df[
    "PSO_Percentage_Error"
] = (

    result_df[
        "PSO_Absolute_Error_mm"
    ]

    /

    result_df[
        "Actual_Rainfall_mm"
    ]

) * 100


result_df = (

    result_df

    .sort_values(
        by="Normal_Rainfall_mm"
    )

    .reset_index(
        drop=True
    )
)


result_df.insert(

    0,

    "Sample_ID",

    np.arange(
        1,
        len(
            result_df
        )
        + 1
    )
)


result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)


print(
    "\nPSO Result CSV Saved:"
)

print(
    RESULT_CSV_PATH
)


# ======================================================================
# 45. PSO RESULT GRAPH
# ======================================================================

plt.figure(
    figsize=(12, 6)
)


plt.plot(

    result_df[
        "Sample_ID"
    ],

    result_df[
        "Actual_Rainfall_mm"
    ],

    linewidth=2,

    label="Actual Rainfall"
)


plt.plot(

    result_df[
        "Sample_ID"
    ],

    result_df[
        "PSO_Predicted_Rainfall_mm"
    ],

    linestyle="--",

    linewidth=2,

    label="PSO Predicted Rainfall"
)


plt.xlabel(
    "Test Sample"
)


plt.ylabel(
    "Rainfall (mm)"
)


plt.title(
    "PSO Actual vs Predicted Rainfall"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nPSO Result Graph Saved:"
)

print(
    RESULT_GRAPH_PATH
)


# ======================================================================
# 46. GENERATE PSO PREDICTION RANGE
# ======================================================================

prediction_normal_values = np.linspace(

    normal_rainfall * 0.70,

    normal_rainfall * 1.30,

    50
)


prediction_input = (
    prediction_normal_values
    .reshape(
        -1,
        1
    )
)


prediction_input_scaled = (
    feature_scaler
    .transform(
        prediction_input
    )
)


prediction_scaled = (
    final_model.predict(
        prediction_input_scaled,
        verbose=0
    )
)


prediction_values = (

    target_scaler

    .inverse_transform(
        prediction_scaled
    )

    .flatten()
)


# ======================================================================
# 47. PSO PREDICTION CSV
# ======================================================================

prediction_df = pd.DataFrame(
    {

        "Prediction_ID":
            np.arange(
                1,
                len(
                    prediction_normal_values
                )
                + 1
            ),

        "Normal_Rainfall_mm":
            prediction_normal_values,

        "PSO_Predicted_Actual_Rainfall_mm":
            prediction_values

    }
)


prediction_df[
    "PSO_Predicted_Difference_mm"
] = (

    prediction_df[
        "PSO_Predicted_Actual_Rainfall_mm"
    ]

    -

    prediction_df[
        "Normal_Rainfall_mm"
    ]
)


prediction_df[
    "PSO_Predicted_Deviation_Percentage"
] = (

    prediction_df[
        "PSO_Predicted_Difference_mm"
    ]

    /

    prediction_df[
        "Normal_Rainfall_mm"
    ]

) * 100


prediction_df[
    "PSO_Rainfall_Category"
] = np.where(

    prediction_df[
        "PSO_Predicted_Deviation_Percentage"
    ] > 0,

    "Above Normal",

    np.where(

        prediction_df[
            "PSO_Predicted_Deviation_Percentage"
        ] < 0,

        "Below Normal",

        "Normal"
    )
)


prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)


print(
    "\nPSO Prediction CSV Saved:"
)

print(
    PREDICTION_CSV_PATH
)


# ======================================================================
# 48. PSO PREDICTION GRAPH
# ======================================================================

plt.figure(
    figsize=(11, 6)
)


plt.plot(

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    prediction_df[
        "PSO_Predicted_Actual_Rainfall_mm"
    ],

    marker="o",

    markersize=4,

    linewidth=2,

    label="PSO Predicted Rainfall"
)


plt.plot(

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    linestyle="--",

    linewidth=2,

    label="Normal Rainfall Reference"
)


plt.scatter(

    [
        normal_rainfall
    ],

    [
        actual_rainfall
    ],

    s=100,

    label="Observed Rainfall",

    zorder=5
)


plt.xlabel(
    "Normal Rainfall (mm)"
)


plt.ylabel(
    "Rainfall (mm)"
)


plt.title(
    "PSO Rainfall Prediction Graph"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nPSO Prediction Graph Saved:"
)

print(
    PREDICTION_GRAPH_PATH
)


# ======================================================================
# 49. SAVE PSO YAML
# ======================================================================

yaml_config = {

    "project": {

        "name":
            "AI-Based Rainfall Pattern Analysis and Prediction",

        "optimization":
            "Particle Swarm Optimization",

        "model":
            "Artificial Neural Network",

        "version":
            "1.0"

    },

    "dataset": {

        "filename":
            os.path.basename(
                DATASET_PATH
            ),

        "path":
            DATASET_PATH,

        "rows":
            int(
                df.shape[0]
            ),

        "columns":
            int(
                df.shape[1]
            ),

        "description_column":
            description_column,

        "rainfall_column":
            rainfall_column

    },

    "rainfall_analysis": {

        "normal_rainfall_mm":
            float(
                normal_rainfall
            ),

        "actual_rainfall_mm":
            float(
                actual_rainfall
            ),

        "difference_mm":
            float(
                difference
            ),

        "deviation_percentage":
            float(
                percentage_deviation
            ),

        "pattern":
            rainfall_pattern

    },

    "pso": {

        "swarm_size":
            SWARM_SIZE,

        "iterations":
            PSO_ITERATIONS,

        "inertia_weight":
            W,

        "cognitive_coefficient":
            C1,

        "social_coefficient":
            C2,

        "best_neurons_1":
            BEST_NEURONS_1,

        "best_neurons_2":
            BEST_NEURONS_2,

        "best_learning_rate":
            float(
                BEST_LEARNING_RATE
            ),

        "best_batch_size":
            BEST_BATCH_SIZE,

        "best_validation_loss":
            float(
                global_best_score
            )

    },

    "model_performance": {

        "mae":
            float(
                mae
            ),

        "mse":
            float(
                mse
            ),

        "rmse":
            float(
                rmse
            ),

        "r2_score":
            float(
                r2
            ),

        "mape":
            float(
                mape
            ),

        "regression_accuracy":
            float(
                regression_accuracy
            )

    }

}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as yaml_file:

    yaml.safe_dump(

        yaml_config,

        yaml_file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "\nPSO YAML Saved:"
)

print(
    YAML_PATH
)


# ======================================================================
# 50. SAVE PSO JSON METADATA
# ======================================================================

json_metadata = {

    "project_name":
        "AI-Based Rainfall Pattern Analysis and Prediction",

    "optimization_algorithm":
        "Particle Swarm Optimization",

    "model":
        "Artificial Neural Network Regression",

    "dataset":
        os.path.basename(
            DATASET_PATH
        ),

    "rainfall_statistics": {

        "normal_rainfall_mm":
            round(
                normal_rainfall,
                4
            ),

        "actual_rainfall_mm":
            round(
                actual_rainfall,
                4
            ),

        "difference_mm":
            round(
                difference,
                4
            ),

        "deviation_percentage":
            round(
                percentage_deviation,
                4
            ),

        "pattern":
            rainfall_pattern

    },

    "pso_configuration": {

        "swarm_size":
            SWARM_SIZE,

        "iterations":
            PSO_ITERATIONS,

        "inertia_weight":
            W,

        "cognitive_coefficient":
            C1,

        "social_coefficient":
            C2

    },

    "best_pso_parameters": {

        "neurons_1":
            BEST_NEURONS_1,

        "neurons_2":
            BEST_NEURONS_2,

        "learning_rate":
            BEST_LEARNING_RATE,

        "batch_size":
            BEST_BATCH_SIZE,

        "validation_loss":
            global_best_score

    },

    "model_performance": {

        "mae":
            round(
                mae,
                6
            ),

        "mse":
            round(
                mse,
                6
            ),

        "rmse":
            round(
                rmse,
                6
            ),

        "r2":
            round(
                r2,
                6
            ),

        "mape":
            round(
                mape,
                6
            ),

        "regression_accuracy":
            round(
                regression_accuracy,
                6
            )

    },

    "original_prediction": {

        "normal_rainfall_mm":
            round(
                normal_rainfall,
                4
            ),

        "actual_rainfall_mm":
            round(
                actual_rainfall,
                4
            ),

        "pso_predicted_rainfall_mm":
            round(
                predicted_actual_rainfall,
                4
            ),

        "absolute_error_mm":
            round(
                prediction_error,
                6
            ),

        "percentage_error":
            round(
                percentage_error,
                6
            )

    },

    "generated_files": [

        "pso_rainfall_prediction_model.h5",

        "pso_rainfall_preprocessor.pkl",

        "pso_rainfall_config.yaml",

        "pso_rainfall_metadata.json",

        "pso_accuracy_graph.png",

        "pso_heatmap.png",

        "pso_comparison_graph.png",

        "pso_result.csv",

        "pso_result_graph.png",

        "pso_prediction.csv",

        "pso_prediction_graph.png",

        "pso_optimization_history.csv",

        "pso_optimization_graph.png"

    ]

}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as json_file:

    json.dump(

        json_metadata,

        json_file,

        indent=4,

        ensure_ascii=False
    )


print(
    "\nPSO JSON Saved:"
)

print(
    JSON_PATH
)


# ======================================================================
# 51. VERIFY SAVED PSO MODEL
# ======================================================================
#
# IMPORTANT KERAS H5 FIX:
#
# compile=False prevents the legacy H5 compile configuration from
# producing errors such as:
#
# TypeError: Could not locate function 'mse'
#
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "VERIFYING SAVED PSO MODEL"
)

print(
    "=" * 78
)


loaded_model = (
    tf.keras.models.load_model(
        MODEL_PATH,
        compile=False
    )
)


loaded_preprocessor = (
    joblib.load(
        PKL_PATH
    )
)


loaded_feature_scaler = (
    loaded_preprocessor[
        "feature_scaler"
    ]
)


loaded_target_scaler = (
    loaded_preprocessor[
        "target_scaler"
    ]
)


verify_input = np.array(
    [
        [
            normal_rainfall
        ]
    ],
    dtype=np.float32
)


verify_scaled = (
    loaded_feature_scaler
    .transform(
        verify_input
    )
)


verify_prediction_scaled = (
    loaded_model.predict(
        verify_scaled,
        verbose=0
    )
)


verify_prediction = float(

    loaded_target_scaler

    .inverse_transform(
        verify_prediction_scaled
    )[0][0]
)


print(
    "\nPSO H5 model loaded successfully."
)


print(
    f"Verification Prediction: "
    f"{verify_prediction:.2f} mm"
)


# ======================================================================
# 52. CUSTOM PSO PREDICTION FUNCTION
# ======================================================================

def pso_predict_actual_rainfall(
    normal_rainfall_input
):

    """
    Predict actual rainfall using the saved PSO-optimized ANN.
    """

    model_loaded = (
        tf.keras.models.load_model(
            MODEL_PATH,
            compile=False
        )
    )


    preprocessing_loaded = (
        joblib.load(
            PKL_PATH
        )
    )


    x_scaler = (
        preprocessing_loaded[
            "feature_scaler"
        ]
    )


    y_scaler = (
        preprocessing_loaded[
            "target_scaler"
        ]
    )


    input_data = np.array(
        [
            [
                float(
                    normal_rainfall_input
                )
            ]
        ],
        dtype=np.float32
    )


    input_scaled = (
        x_scaler
        .transform(
            input_data
        )
    )


    prediction_scaled = (
        model_loaded.predict(
            input_scaled,
            verbose=0
        )
    )


    prediction = (
        y_scaler
        .inverse_transform(
            prediction_scaled
        )
    )


    return float(
        prediction[0][0]
    )


# ======================================================================
# 53. EXAMPLE CUSTOM PSO PREDICTION
# ======================================================================

example_normal_rainfall = (
    1000.0
)


example_prediction = (
    pso_predict_actual_rainfall(
        example_normal_rainfall
    )
)


print(
    "\n" + "=" * 78
)

print(
    "PSO CUSTOM PREDICTION"
)

print(
    "=" * 78
)


print(
    f"\nNormal Rainfall Input : "
    f"{example_normal_rainfall:.2f} mm"
)


print(
    f"PSO Predicted Rainfall: "
    f"{example_prediction:.2f} mm"
)


# ======================================================================
# 54. RESULT CSV PREVIEW
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "PSO RESULT CSV PREVIEW"
)

print(
    "=" * 78
)


print(

    result_df

    .head(
        10
    )

    .to_string(
        index=False
    )
)


# ======================================================================
# 55. PREDICTION CSV PREVIEW
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "PSO PREDICTION CSV PREVIEW"
)

print(
    "=" * 78
)


print(

    prediction_df

    .head(
        10
    )

    .to_string(
        index=False
    )
)


# ======================================================================
# 56. VERIFY ALL PSO OUTPUT FILES
# ======================================================================

all_pso_files = [

    MODEL_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    PSO_HISTORY_CSV_PATH,

    PSO_OPTIMIZATION_GRAPH_PATH

]


print(
    "\n" + "=" * 78
)

print(
    "VERIFYING ALL PSO OUTPUT FILES"
)

print(
    "=" * 78
)


for file_path in all_pso_files:

    if os.path.exists(
        file_path
    ):

        file_size_kb = (

            os.path.getsize(
                file_path
            )

            /

            1024
        )


        print(

            f"[OK] "

            f"{os.path.basename(file_path)} "

            f"({file_size_kb:.2f} KB)"
        )

    else:

        print(

            f"[ERROR] "

            f"{os.path.basename(file_path)}"
        )


# ======================================================================
# 57. FINAL SUMMARY
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "PSO PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 78
)


print(
    f"""
PROJECT
-------
AI-Based Rainfall Pattern Analysis and Prediction


OPTIMIZATION
------------
Particle Swarm Optimization (PSO)


DATASET
-------
{DATASET_PATH}


RAINFALL ANALYSIS
-----------------
Normal Rainfall:
{normal_rainfall:.2f} mm

Actual Rainfall:
{actual_rainfall:.2f} mm

Difference:
{difference:.2f} mm

Deviation:
{percentage_deviation:.4f}%

Pattern:
{rainfall_pattern}


BEST PSO PARAMETERS
-------------------
First Hidden Layer:
{BEST_NEURONS_1}

Second Hidden Layer:
{BEST_NEURONS_2}

Learning Rate:
{BEST_LEARNING_RATE:.8f}

Batch Size:
{BEST_BATCH_SIZE}

Best Validation Loss:
{global_best_score:.8f}


PSO MODEL PERFORMANCE
---------------------
MAE:
{mae:.6f}

MSE:
{mse:.6f}

RMSE:
{rmse:.6f}

R2 Score:
{r2:.6f}

MAPE:
{mape:.6f}%

Regression Accuracy:
{regression_accuracy:.6f}%


ORIGINAL DATA PREDICTION
------------------------
Normal Rainfall:
{normal_rainfall:.2f} mm

Actual Rainfall:
{actual_rainfall:.2f} mm

PSO Predicted Rainfall:
{predicted_actual_rainfall:.2f} mm


GENERATED PSO FILES
-------------------
1. pso_rainfall_prediction_model.h5

2. pso_rainfall_preprocessor.pkl

3. pso_rainfall_config.yaml

4. pso_rainfall_metadata.json

5. pso_accuracy_graph.png

6. pso_heatmap.png

7. pso_comparison_graph.png

8. pso_result.csv

9. pso_result_graph.png

10. pso_prediction.csv

11. pso_prediction_graph.png

12. pso_optimization_history.csv

13. pso_optimization_graph.png


H5 COMPATIBILITY FIX
---------------------
Model compilation uses:

tf.keras.losses.MeanSquaredError()

Model loading uses:

tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

Therefore, the previous legacy H5
'mse' deserialization error is avoided.


OUTPUT DIRECTORY
----------------
{PROJECT_DIR}
"""
)


print(
    "=" * 78
)
```
