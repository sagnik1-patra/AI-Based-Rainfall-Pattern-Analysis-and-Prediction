```python
import os
import json
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

import joblib
import yaml

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Input,
    Dense
)

from tensorflow.keras.callbacks import (
    EarlyStopping
)


# ======================================================================
# 2. SETTINGS
# ======================================================================

warnings.filterwarnings(
    "ignore"
)

np.random.seed(
    42
)

tf.random.set_seed(
    42
)


# ======================================================================
# 3. PROJECT PATH
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Rainfall Pattern Analysis and Prediction"
)


DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_june_2019_to_may_2020.csv"
)


# ======================================================================
# 4. MODEL FILE PATHS
# ======================================================================

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_prediction_model.h5"
)


PKL_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_preprocessor.pkl"
)


YAML_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_config.yaml"
)


JSON_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_metadata.json"
)


# ======================================================================
# 5. GRAPH PATHS
# ======================================================================

ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "accuracy_graph.png"
)


HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "heatmap.png"
)


COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "comparison_graph.png"
)


RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "result_graph.png"
)


PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "prediction_graph.png"
)


# ======================================================================
# 6. CSV PATHS
# ======================================================================

RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "result.csv"
)


PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "prediction.csv"
)


# ======================================================================
# 7. CREATE PROJECT DIRECTORY
# ======================================================================

os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ======================================================================
# 8. DISPLAY PROJECT INFORMATION
# ======================================================================

print(
    "=" * 75
)

print(
    "AI-BASED RAINFALL PATTERN ANALYSIS AND PREDICTION"
)

print(
    "=" * 75
)


print(
    "\nProject Directory:"
)

print(
    PROJECT_DIR
)


print(
    "\nDataset:"
)

print(
    DATASET_PATH
)


# ======================================================================
# 9. CHECK DATASET
# ======================================================================

if not os.path.exists(
    DATASET_PATH
):

    raise FileNotFoundError(
        f"\nDataset not found:\n"
        f"{DATASET_PATH}"
    )


print(
    "\nDataset found successfully."
)


# ======================================================================
# 10. LOAD DATASET
# ======================================================================

df = pd.read_csv(
    DATASET_PATH
)


# Clean column names
df.columns = (
    df.columns
    .str.strip()
)


print(
    "\n" + "=" * 75
)

print(
    "DATASET LOADED"
)

print(
    "=" * 75
)


print(
    "\nDataset Shape:",
    df.shape
)


print(
    "\nColumns:"
)

for column in df.columns:

    print(
        "-",
        column
    )


print(
    "\nOriginal Dataset:"
)

print(
    df.to_string(
        index=False
    )
)


# ======================================================================
# 11. IDENTIFY DESCRIPTION AND RAINFALL COLUMNS
# ======================================================================

description_column = None

rainfall_column = None


for column in df.columns:

    column_lower = (
        column.lower()
    )

    if "description" in column_lower:

        description_column = column

    if "rainfall" in column_lower:

        rainfall_column = column


# ----------------------------------------------------------------------
# Fallback for description
# ----------------------------------------------------------------------

if description_column is None:

    description_column = (
        df.columns[0]
    )


# ----------------------------------------------------------------------
# Fallback for rainfall
# ----------------------------------------------------------------------

if rainfall_column is None:

    numeric_columns = []

    for column in df.columns:

        converted = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        if (
            converted
            .notna()
            .sum()
            > 0
        ):

            numeric_columns.append(
                column
            )

    if len(
        numeric_columns
    ) > 0:

        rainfall_column = (
            numeric_columns[0]
        )


if rainfall_column is None:

    raise ValueError(
        "Rainfall column could not be identified."
    )


print(
    "\nDescription Column:",
    description_column
)


print(
    "Rainfall Column:",
    rainfall_column
)


# ======================================================================
# 12. CLEAN DATASET
# ======================================================================

df[
    description_column
] = (
    df[
        description_column
    ]
    .astype(str)
    .str.strip()
)


df[
    rainfall_column
] = pd.to_numeric(
    df[
        rainfall_column
    ],
    errors="coerce"
)


df = (
    df
    .dropna(
        subset=[
            rainfall_column
        ]
    )
    .reset_index(
        drop=True
    )
)


# ======================================================================
# 13. FIND NORMAL RAINFALL
# ======================================================================

normal_rows = df[
    df[
        description_column
    ]
    .str.lower()
    .str.contains(
        "normal",
        na=False
    )
]


if len(
    normal_rows
) == 0:

    raise ValueError(
        "Normal Rainfall row was not found."
    )


normal_rainfall = float(
    normal_rows[
        rainfall_column
    ].iloc[0]
)


# ======================================================================
# 14. FIND ACTUAL RAINFALL
# ======================================================================

actual_rows = df[
    df[
        description_column
    ]
    .str.lower()
    .str.contains(
        "actual",
        na=False
    )
]


if len(
    actual_rows
) == 0:

    raise ValueError(
        "Actual Rainfall row was not found."
    )


actual_rainfall = float(
    actual_rows[
        rainfall_column
    ].iloc[0]
)


# ======================================================================
# 15. BASIC ANALYSIS
# ======================================================================

difference = (
    actual_rainfall
    -
    normal_rainfall
)


percentage_deviation = (
    difference
    /
    normal_rainfall
) * 100


rainfall_ratio = (
    actual_rainfall
    /
    normal_rainfall
)


if percentage_deviation > 0:

    rainfall_pattern = (
        "Above Normal"
    )

elif percentage_deviation < 0:

    rainfall_pattern = (
        "Below Normal"
    )

else:

    rainfall_pattern = (
        "Normal"
    )


print(
    "\n" + "=" * 75
)

print(
    "RAINFALL ANALYSIS"
)

print(
    "=" * 75
)


print(
    f"\nNormal Rainfall      : "
    f"{normal_rainfall:.2f} mm"
)


print(
    f"Actual Rainfall      : "
    f"{actual_rainfall:.2f} mm"
)


print(
    f"Difference           : "
    f"{difference:.2f} mm"
)


print(
    f"Percentage Deviation : "
    f"{percentage_deviation:.4f}%"
)


print(
    f"Rainfall Ratio       : "
    f"{rainfall_ratio:.6f}"
)


print(
    f"Rainfall Pattern     : "
    f"{rainfall_pattern}"
)


# ======================================================================
# 16. CREATE TRAINING DATA
# ======================================================================
#
# NOTE:
#
# The original CSV contains only an aggregate Normal Rainfall and
# Actual Rainfall pair.
#
# A synthetic local dataset is therefore created around this relationship
# to demonstrate the ANN training, prediction, graph generation and
# serialization pipeline.
#
# ======================================================================

TOTAL_SAMPLES = 500


training_normal_values = np.linspace(
    normal_rainfall * 0.60,
    normal_rainfall * 1.40,
    TOTAL_SAMPLES
)


# Add a small deterministic variation
variation = (
    1
    +
    0.015
    *
    np.sin(
        np.linspace(
            0,
            8 * np.pi,
            TOTAL_SAMPLES
        )
    )
)


training_actual_values = (
    training_normal_values
    *
    rainfall_ratio
    *
    variation
)


X = (
    training_normal_values
    .reshape(
        -1,
        1
    )
)


y = (
    training_actual_values
    .reshape(
        -1,
        1
    )
)


print(
    "\n" + "=" * 75
)

print(
    "TRAINING DATA"
)

print(
    "=" * 75
)


print(
    "\nTotal Samples:",
    len(X)
)


# ======================================================================
# 17. TRAIN TEST SPLIT
# ======================================================================

(
    X_train,
    X_test,
    y_train,
    y_test
) = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42
)


print(
    "Training Samples:",
    len(X_train)
)


print(
    "Testing Samples:",
    len(X_test)
)


# ======================================================================
# 18. CREATE SCALERS
# ======================================================================

feature_scaler = (
    StandardScaler()
)


target_scaler = (
    StandardScaler()
)


# ======================================================================
# 19. SCALE TRAINING DATA
# ======================================================================

X_train_scaled = (
    feature_scaler
    .fit_transform(
        X_train
    )
)


X_test_scaled = (
    feature_scaler
    .transform(
        X_test
    )
)


y_train_scaled = (
    target_scaler
    .fit_transform(
        y_train
    )
)


y_test_scaled = (
    target_scaler
    .transform(
        y_test
    )
)


# ======================================================================
# 20. BUILD ANN MODEL
# ======================================================================

model = Sequential(
    [

        Input(
            shape=(1,)
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            8,
            activation="relu"
        ),

        Dense(
            1,
            activation="linear"
        )

    ]
)


# ======================================================================
# 21. COMPILE MODEL
# ======================================================================
#
# CRITICAL FIX:
#
# DO NOT USE:
#
# loss="mse"
# metrics=["mae"]
#
# Older HDF5 serialization may store "mse" as a function reference.
# Newer Keras versions can fail while deserializing it.
#
# Use explicit serializable Keras objects instead.
# ======================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss=tf.keras.losses.MeanSquaredError(),

    metrics=[
        tf.keras.metrics.MeanAbsoluteError(
            name="mae"
        )
    ]
)


print(
    "\n" + "=" * 75
)

print(
    "MODEL ARCHITECTURE"
)

print(
    "=" * 75
)


model.summary()


# ======================================================================
# 22. EARLY STOPPING
# ======================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=40,

    restore_best_weights=True,

    verbose=0
)


# ======================================================================
# 23. TRAIN MODEL
# ======================================================================

print(
    "\n" + "=" * 75
)

print(
    "TRAINING MODEL"
)

print(
    "=" * 75
)


history = model.fit(

    X_train_scaled,

    y_train_scaled,

    validation_split=0.20,

    epochs=500,

    batch_size=16,

    callbacks=[
        early_stopping
    ],

    verbose=0
)


epochs_completed = len(
    history.history[
        "loss"
    ]
)


print(
    "\nTraining completed successfully."
)


print(
    "Epochs Completed:",
    epochs_completed
)


# ======================================================================
# 24. TEST SET PREDICTIONS
# ======================================================================

test_prediction_scaled = (
    model.predict(
        X_test_scaled,
        verbose=0
    )
)


test_predictions = (
    target_scaler
    .inverse_transform(
        test_prediction_scaled
    )
)


actual_test_values = (
    y_test.flatten()
)


predicted_test_values = (
    test_predictions.flatten()
)


# ======================================================================
# 25. CALCULATE METRICS
# ======================================================================

mae = mean_absolute_error(
    actual_test_values,
    predicted_test_values
)


mse = mean_squared_error(
    actual_test_values,
    predicted_test_values
)


rmse = np.sqrt(
    mse
)


r2 = r2_score(
    actual_test_values,
    predicted_test_values
)


safe_actual = np.where(
    actual_test_values == 0,
    1e-8,
    actual_test_values
)


mape = (
    np.mean(
        np.abs(
            (
                actual_test_values
                -
                predicted_test_values
            )
            /
            safe_actual
        )
    )
    *
    100
)


regression_accuracy = max(
    0.0,
    100.0 - mape
)


print(
    "\n" + "=" * 75
)

print(
    "MODEL PERFORMANCE"
)

print(
    "=" * 75
)


print(
    f"\nMAE                 : "
    f"{mae:.6f}"
)


print(
    f"MSE                 : "
    f"{mse:.6f}"
)


print(
    f"RMSE                : "
    f"{rmse:.6f}"
)


print(
    f"R2 Score            : "
    f"{r2:.6f}"
)


print(
    f"MAPE                : "
    f"{mape:.6f}%"
)


print(
    f"Regression Accuracy : "
    f"{regression_accuracy:.6f}%"
)


# ======================================================================
# 26. ORIGINAL RAINFALL PREDICTION
# ======================================================================

original_input = np.array(
    [
        [
            normal_rainfall
        ]
    ],
    dtype=np.float32
)


original_input_scaled = (
    feature_scaler
    .transform(
        original_input
    )
)


original_prediction_scaled = (
    model.predict(
        original_input_scaled,
        verbose=0
    )
)


predicted_actual_rainfall = float(
    target_scaler
    .inverse_transform(
        original_prediction_scaled
    )[0][0]
)


prediction_error = abs(
    actual_rainfall
    -
    predicted_actual_rainfall
)


percentage_error = (
    prediction_error
    /
    actual_rainfall
) * 100


print(
    "\n" + "=" * 75
)

print(
    "ORIGINAL RAINFALL PREDICTION"
)

print(
    "=" * 75
)


print(
    f"\nNormal Rainfall    : "
    f"{normal_rainfall:.2f} mm"
)


print(
    f"Actual Rainfall    : "
    f"{actual_rainfall:.2f} mm"
)


print(
    f"Predicted Rainfall : "
    f"{predicted_actual_rainfall:.2f} mm"
)


print(
    f"Absolute Error     : "
    f"{prediction_error:.4f} mm"
)


print(
    f"Percentage Error   : "
    f"{percentage_error:.4f}%"
)


# ======================================================================
# 27. SAVE H5 MODEL
# ======================================================================
#
# H5 is saved because it was specifically requested.
#
# ======================================================================

print(
    "\n" + "=" * 75
)

print(
    "SAVING MODEL"
)

print(
    "=" * 75
)


model.save(
    MODEL_PATH
)


print(
    "\nH5 Model Saved:"
)

print(
    MODEL_PATH
)


# ======================================================================
# 28. SAVE PREPROCESSOR PKL
# ======================================================================

preprocessor_data = {

    "feature_scaler":
        feature_scaler,

    "target_scaler":
        target_scaler,

    "feature_name":
        "Normal Rainfall",

    "target_name":
        "Actual Rainfall",

    "description_column":
        description_column,

    "rainfall_column":
        rainfall_column,

    "normal_rainfall":
        normal_rainfall,

    "actual_rainfall":
        actual_rainfall,

    "rainfall_ratio":
        rainfall_ratio

}


joblib.dump(
    preprocessor_data,
    PKL_PATH
)


print(
    "\nPKL Saved:"
)

print(
    PKL_PATH
)


# ======================================================================
# 29. ACCURACY GRAPH
# ======================================================================
#
# For regression, conventional classification accuracy does not apply.
#
# This graph visualizes a learning score derived from model loss.
# ======================================================================

print(
    "\nGenerating Accuracy Graph..."
)


training_loss = np.array(
    history.history[
        "loss"
    ]
)


validation_loss = np.array(
    history.history[
        "val_loss"
    ]
)


training_score = (
    100.0
    /
    (
        1.0
        +
        training_loss
    )
)


validation_score = (
    100.0
    /
    (
        1.0
        +
        validation_loss
    )
)


epoch_numbers = np.arange(
    1,
    epochs_completed + 1
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(
    epoch_numbers,
    training_score,
    linewidth=2,
    label="Training Score"
)


plt.plot(
    epoch_numbers,
    validation_score,
    linewidth=2,
    label="Validation Score"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Learning Score (%)"
)


plt.title(
    "Rainfall Prediction Model Accuracy Graph"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Accuracy Graph Saved:"
)

print(
    ACCURACY_GRAPH_PATH
)


# ======================================================================
# 30. CREATE CORRELATION DATA
# ======================================================================

correlation_df = pd.DataFrame(
    {

        "Normal Rainfall":
            training_normal_values,

        "Actual Rainfall":
            training_actual_values,

        "Difference":
            (
                training_actual_values
                -
                training_normal_values
            ),

        "Ratio":
            (
                training_actual_values
                /
                training_normal_values
            ),

        "Deviation":
            (
                (
                    training_actual_values
                    -
                    training_normal_values
                )
                /
                training_normal_values
            )
            * 100

    }
)


correlation_matrix = (
    correlation_df
    .corr()
)


# ======================================================================
# 31. HEATMAP
# ======================================================================

print(
    "\nGenerating Heatmap..."
)


plt.figure(
    figsize=(10, 8)
)


image = plt.imshow(

    correlation_matrix,

    cmap="coolwarm",

    interpolation="nearest",

    vmin=-1,

    vmax=1
)


plt.colorbar(
    image,
    label="Correlation"
)


heatmap_labels = (
    correlation_matrix
    .columns
    .tolist()
)


plt.xticks(
    range(
        len(
            heatmap_labels
        )
    ),
    heatmap_labels,
    rotation=45,
    ha="right"
)


plt.yticks(
    range(
        len(
            heatmap_labels
        )
    ),
    heatmap_labels
)


for i in range(
    len(
        heatmap_labels
    )
):

    for j in range(
        len(
            heatmap_labels
        )
    ):

        correlation_value = (
            correlation_matrix
            .iloc[
                i,
                j
            ]
        )

        plt.text(
            j,
            i,
            f"{correlation_value:.2f}",
            ha="center",
            va="center"
        )


plt.title(
    "Rainfall Feature Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Heatmap Saved:"
)

print(
    HEATMAP_PATH
)


# ======================================================================
# 32. COMPARISON GRAPH
# ======================================================================

print(
    "\nGenerating Comparison Graph..."
)


comparison_labels = [
    "Regression Accuracy",
    "R2 Score",
    "MAPE"
]


comparison_values = [

    regression_accuracy,

    max(
        0.0,
        r2 * 100.0
    ),

    mape

]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(
    comparison_labels,
    comparison_values
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "Rainfall Prediction Model Performance Comparison"
)


for bar, value in zip(
    bars,
    comparison_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        bar.get_height(),

        f"{value:.2f}",

        ha="center",

        va="bottom",

        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Comparison Graph Saved:"
)

print(
    COMPARISON_GRAPH_PATH
)


# ======================================================================
# 33. CREATE RESULT CSV
# ======================================================================

result_df = pd.DataFrame(
    {

        "Normal_Rainfall_mm":
            X_test.flatten(),

        "Actual_Rainfall_mm":
            actual_test_values,

        "Predicted_Rainfall_mm":
            predicted_test_values

    }
)


result_df[
    "Absolute_Error_mm"
] = np.abs(

    result_df[
        "Actual_Rainfall_mm"
    ]

    -

    result_df[
        "Predicted_Rainfall_mm"
    ]
)


result_df[
    "Percentage_Error"
] = (

    result_df[
        "Absolute_Error_mm"
    ]

    /

    result_df[
        "Actual_Rainfall_mm"
    ]

) * 100


result_df = (
    result_df
    .sort_values(
        by="Normal_Rainfall_mm"
    )
    .reset_index(
        drop=True
    )
)


result_df.insert(

    0,

    "Sample_ID",

    np.arange(
        1,
        len(
            result_df
        )
        + 1
    )
)


result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)


print(
    "\nResult CSV Saved:"
)

print(
    RESULT_CSV_PATH
)


# ======================================================================
# 34. RESULT GRAPH
# ======================================================================

print(
    "\nGenerating Result Graph..."
)


plt.figure(
    figsize=(12, 6)
)


plt.plot(

    result_df[
        "Sample_ID"
    ],

    result_df[
        "Actual_Rainfall_mm"
    ],

    linewidth=2,

    label="Actual Rainfall"
)


plt.plot(

    result_df[
        "Sample_ID"
    ],

    result_df[
        "Predicted_Rainfall_mm"
    ],

    linestyle="--",

    linewidth=2,

    label="Predicted Rainfall"
)


plt.xlabel(
    "Test Sample"
)


plt.ylabel(
    "Rainfall (mm)"
)


plt.title(
    "Actual vs Predicted Rainfall Results"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Result Graph Saved:"
)

print(
    RESULT_GRAPH_PATH
)


# ======================================================================
# 35. CREATE PREDICTION RANGE
# ======================================================================

prediction_normal_values = np.linspace(

    normal_rainfall * 0.70,

    normal_rainfall * 1.30,

    50
)


prediction_input = (
    prediction_normal_values
    .reshape(
        -1,
        1
    )
)


prediction_input_scaled = (
    feature_scaler
    .transform(
        prediction_input
    )
)


prediction_scaled = (
    model.predict(
        prediction_input_scaled,
        verbose=0
    )
)


prediction_values = (
    target_scaler
    .inverse_transform(
        prediction_scaled
    )
    .flatten()
)


# ======================================================================
# 36. CREATE PREDICTION CSV
# ======================================================================

prediction_df = pd.DataFrame(
    {

        "Prediction_ID":
            np.arange(
                1,
                len(
                    prediction_normal_values
                )
                + 1
            ),

        "Normal_Rainfall_mm":
            prediction_normal_values,

        "Predicted_Actual_Rainfall_mm":
            prediction_values

    }
)


prediction_df[
    "Predicted_Difference_mm"
] = (

    prediction_df[
        "Predicted_Actual_Rainfall_mm"
    ]

    -

    prediction_df[
        "Normal_Rainfall_mm"
    ]
)


prediction_df[
    "Predicted_Deviation_Percentage"
] = (

    prediction_df[
        "Predicted_Difference_mm"
    ]

    /

    prediction_df[
        "Normal_Rainfall_mm"
    ]

) * 100


prediction_df[
    "Rainfall_Category"
] = np.where(

    prediction_df[
        "Predicted_Deviation_Percentage"
    ] > 0,

    "Above Normal",

    np.where(

        prediction_df[
            "Predicted_Deviation_Percentage"
        ] < 0,

        "Below Normal",

        "Normal"
    )
)


prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)


print(
    "\nPrediction CSV Saved:"
)

print(
    PREDICTION_CSV_PATH
)


# ======================================================================
# 37. PREDICTION GRAPH
# ======================================================================

print(
    "\nGenerating Prediction Graph..."
)


plt.figure(
    figsize=(11, 6)
)


plt.plot(

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    prediction_df[
        "Predicted_Actual_Rainfall_mm"
    ],

    marker="o",

    markersize=4,

    linewidth=2,

    label="Predicted Actual Rainfall"
)


plt.plot(

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    linestyle="--",

    linewidth=2,

    label="Normal Rainfall Reference"
)


plt.scatter(

    [
        normal_rainfall
    ],

    [
        actual_rainfall
    ],

    s=100,

    label="Observed Rainfall",

    zorder=5
)


plt.xlabel(
    "Normal Rainfall (mm)"
)


plt.ylabel(
    "Rainfall (mm)"
)


plt.title(
    "Rainfall Prediction Graph"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Prediction Graph Saved:"
)

print(
    PREDICTION_GRAPH_PATH
)


# ======================================================================
# 38. CREATE YAML CONFIGURATION
# ======================================================================

yaml_config = {

    "project": {

        "name":
            "AI-Based Rainfall Pattern Analysis and Prediction",

        "version":
            "1.0",

        "task":
            "Regression"

    },

    "dataset": {

        "filename":
            os.path.basename(
                DATASET_PATH
            ),

        "path":
            DATASET_PATH,

        "rows":
            int(
                df.shape[0]
            ),

        "columns":
            int(
                df.shape[1]
            ),

        "description_column":
            description_column,

        "rainfall_column":
            rainfall_column

    },

    "rainfall_analysis": {

        "normal_rainfall_mm":
            float(
                normal_rainfall
            ),

        "actual_rainfall_mm":
            float(
                actual_rainfall
            ),

        "difference_mm":
            float(
                difference
            ),

        "percentage_deviation":
            float(
                percentage_deviation
            ),

        "pattern":
            rainfall_pattern

    },

    "model": {

        "framework":
            "TensorFlow Keras",

        "model_type":
            "Artificial Neural Network Regression",

        "loss":
            "MeanSquaredError",

        "metric":
            "MeanAbsoluteError",

        "optimizer":
            "Adam",

        "epochs_completed":
            int(
                epochs_completed
            ),

        "training_samples":
            int(
                len(
                    X_train
                )
            ),

        "testing_samples":
            int(
                len(
                    X_test
                )
            )

    },

    "performance": {

        "mae":
            float(
                mae
            ),

        "mse":
            float(
                mse
            ),

        "rmse":
            float(
                rmse
            ),

        "r2":
            float(
                r2
            ),

        "mape":
            float(
                mape
            ),

        "regression_accuracy":
            float(
                regression_accuracy
            )

    }

}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as yaml_file:

    yaml.safe_dump(

        yaml_config,

        yaml_file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "\nYAML Saved:"
)

print(
    YAML_PATH
)


# ======================================================================
# 39. CREATE JSON METADATA
# ======================================================================

json_metadata = {

    "project_name":
        "AI-Based Rainfall Pattern Analysis and Prediction",

    "dataset":
        os.path.basename(
            DATASET_PATH
        ),

    "rainfall_statistics": {

        "normal_rainfall_mm":
            round(
                normal_rainfall,
                4
            ),

        "actual_rainfall_mm":
            round(
                actual_rainfall,
                4
            ),

        "difference_mm":
            round(
                difference,
                4
            ),

        "deviation_percentage":
            round(
                percentage_deviation,
                4
            ),

        "rainfall_pattern":
            rainfall_pattern

    },

    "model_performance": {

        "mae":
            round(
                mae,
                6
            ),

        "mse":
            round(
                mse,
                6
            ),

        "rmse":
            round(
                rmse,
                6
            ),

        "r2":
            round(
                r2,
                6
            ),

        "mape":
            round(
                mape,
                6
            ),

        "regression_accuracy":
            round(
                regression_accuracy,
                6
            )

    },

    "original_prediction": {

        "normal_rainfall_mm":
            round(
                normal_rainfall,
                4
            ),

        "actual_rainfall_mm":
            round(
                actual_rainfall,
                4
            ),

        "predicted_rainfall_mm":
            round(
                predicted_actual_rainfall,
                4
            ),

        "absolute_error_mm":
            round(
                prediction_error,
                6
            ),

        "percentage_error":
            round(
                percentage_error,
                6
            )

    },

    "generated_files": [

        "rainfall_prediction_model.h5",

        "rainfall_preprocessor.pkl",

        "rainfall_config.yaml",

        "rainfall_metadata.json",

        "accuracy_graph.png",

        "heatmap.png",

        "comparison_graph.png",

        "result.csv",

        "result_graph.png",

        "prediction.csv",

        "prediction_graph.png"

    ]

}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as json_file:

    json.dump(

        json_metadata,

        json_file,

        indent=4,

        ensure_ascii=False
    )


print(
    "\nJSON Saved:"
)

print(
    JSON_PATH
)


# ======================================================================
# 40. VERIFY GENERATED FILES
# ======================================================================

print(
    "\n" + "=" * 75
)

print(
    "VERIFYING GENERATED FILES"
)

print(
    "=" * 75
)


generated_files = [

    MODEL_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH

]


for file_path in generated_files:

    if os.path.exists(
        file_path
    ):

        size_kb = (
            os.path.getsize(
                file_path
            )
            /
            1024
        )

        print(
            f"[OK] "
            f"{os.path.basename(file_path)} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            f"[ERROR] "
            f"{os.path.basename(file_path)}"
        )


# ======================================================================
# 41. VERIFY SAVED H5 MODEL
# ======================================================================
#
# MOST IMPORTANT FIX:
#
# compile=False tells Keras:
#
# "Load the model architecture and weights, but DO NOT try to deserialize
# the old H5 training configuration."
#
# This prevents:
#
# TypeError: Could not locate function 'mse'
#
# ======================================================================

print(
    "\n" + "=" * 75
)

print(
    "VERIFYING SAVED MODEL"
)

print(
    "=" * 75
)


try:

    loaded_model = (
        tf.keras.models.load_model(
            MODEL_PATH,
            compile=False
        )
    )


    print(
        "\nH5 model loaded successfully "
        "with compile=False."
    )


except Exception as error:

    print(
        "\nModel loading failed:"
    )

    print(
        error
    )

    raise


# ======================================================================
# 42. LOAD PREPROCESSOR
# ======================================================================

loaded_preprocessor = (
    joblib.load(
        PKL_PATH
    )
)


loaded_feature_scaler = (
    loaded_preprocessor[
        "feature_scaler"
    ]
)


loaded_target_scaler = (
    loaded_preprocessor[
        "target_scaler"
    ]
)


# ======================================================================
# 43. VERIFY MODEL PREDICTION
# ======================================================================

verify_input = np.array(
    [
        [
            normal_rainfall
        ]
    ],
    dtype=np.float32
)


verify_scaled = (
    loaded_feature_scaler
    .transform(
        verify_input
    )
)


verify_prediction_scaled = (
    loaded_model.predict(
        verify_scaled,
        verbose=0
    )
)


verify_prediction = float(

    loaded_target_scaler
    .inverse_transform(
        verify_prediction_scaled
    )[0][0]
)


print(
    f"\nVerification Input:"
    f" {normal_rainfall:.2f} mm"
)


print(
    f"Verification Prediction:"
    f" {verify_prediction:.2f} mm"
)


print(
    "\nSaved H5 model verification successful."
)


# ======================================================================
# 44. CUSTOM PREDICTION FUNCTION
# ======================================================================

def predict_actual_rainfall(
    normal_rainfall_input
):

    """
    Predict actual rainfall from a supplied normal rainfall value.

    IMPORTANT:
    Model is loaded with compile=False to avoid legacy H5
    deserialization problems involving mse/mae.
    """

    # --------------------------------------------------------------
    # Load H5 model WITHOUT compile configuration
    # --------------------------------------------------------------

    model_loaded = (
        tf.keras.models.load_model(
            MODEL_PATH,
            compile=False
        )
    )


    # --------------------------------------------------------------
    # Load preprocessing objects
    # --------------------------------------------------------------

    preprocessing_loaded = (
        joblib.load(
            PKL_PATH
        )
    )


    x_scaler = (
        preprocessing_loaded[
            "feature_scaler"
        ]
    )


    y_scaler = (
        preprocessing_loaded[
            "target_scaler"
        ]
    )


    # --------------------------------------------------------------
    # Prepare input
    # --------------------------------------------------------------

    input_data = np.array(
        [
            [
                float(
                    normal_rainfall_input
                )
            ]
        ],
        dtype=np.float32
    )


    # --------------------------------------------------------------
    # Scale
    # --------------------------------------------------------------

    input_scaled = (
        x_scaler
        .transform(
            input_data
        )
    )


    # --------------------------------------------------------------
    # Predict
    # --------------------------------------------------------------

    prediction_scaled = (
        model_loaded.predict(
            input_scaled,
            verbose=0
        )
    )


    # --------------------------------------------------------------
    # Convert back to rainfall units
    # --------------------------------------------------------------

    prediction = (
        y_scaler
        .inverse_transform(
            prediction_scaled
        )
    )


    return float(
        prediction[0][0]
    )


# ======================================================================
# 45. EXAMPLE CUSTOM PREDICTION
# ======================================================================

example_normal_rainfall = (
    1000.0
)


example_prediction = (
    predict_actual_rainfall(
        example_normal_rainfall
    )
)


print(
    "\n" + "=" * 75
)

print(
    "CUSTOM PREDICTION"
)

print(
    "=" * 75
)


print(
    f"\nNormal Rainfall Input:"
    f" {example_normal_rainfall:.2f} mm"
)


print(
    f"Predicted Actual Rainfall:"
    f" {example_prediction:.2f} mm"
)


# ======================================================================
# 46. RESULT CSV PREVIEW
# ======================================================================

print(
    "\n" + "=" * 75
)

print(
    "RESULT CSV PREVIEW"
)

print(
    "=" * 75
)


print(
    result_df
    .head(
        10
    )
    .to_string(
        index=False
    )
)


# ======================================================================
# 47. PREDICTION CSV PREVIEW
# ======================================================================

print(
    "\n" + "=" * 75
)

print(
    "PREDICTION CSV PREVIEW"
)

print(
    "=" * 75
)


print(
    prediction_df
    .head(
        10
    )
    .to_string(
        index=False
    )
)


# ======================================================================
# 48. FINAL SUMMARY
# ======================================================================

print(
    "\n" + "=" * 75
)

print(
    "PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 75
)


print(
    f"""
PROJECT:
AI-Based Rainfall Pattern Analysis and Prediction

DATASET:
{DATASET_PATH}

ORIGINAL RAINFALL ANALYSIS
--------------------------
Normal Rainfall:
{normal_rainfall:.2f} mm

Actual Rainfall:
{actual_rainfall:.2f} mm

Difference:
{difference:.2f} mm

Deviation:
{percentage_deviation:.4f}%

Pattern:
{rainfall_pattern}


MODEL PERFORMANCE
-----------------
MAE:
{mae:.6f}

MSE:
{mse:.6f}

RMSE:
{rmse:.6f}

R2 Score:
{r2:.6f}

MAPE:
{mape:.6f}%

Regression Accuracy:
{regression_accuracy:.6f}%


ORIGINAL PREDICTION
-------------------
Normal Rainfall:
{normal_rainfall:.2f} mm

Actual Rainfall:
{actual_rainfall:.2f} mm

Predicted Rainfall:
{predicted_actual_rainfall:.2f} mm


GENERATED FILES
---------------
1. rainfall_prediction_model.h5

2. rainfall_preprocessor.pkl

3. rainfall_config.yaml

4. rainfall_metadata.json

5. accuracy_graph.png

6. heatmap.png

7. comparison_graph.png

8. result.csv

9. result_graph.png

10. prediction.csv

11. prediction_graph.png


H5 FIX
------
Model compilation uses:
tf.keras.losses.MeanSquaredError()

Model loading uses:
tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

Therefore the legacy H5 'mse'
deserialization error is avoided.


OUTPUT DIRECTORY
----------------
{PROJECT_DIR}
"""
)


print(
    "=" * 75
)
```
