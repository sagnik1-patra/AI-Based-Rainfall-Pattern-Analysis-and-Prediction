```python
import os
import json
import random
import warnings
import gc

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

import joblib
import yaml

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Input,
    Dense
)

from tensorflow.keras.callbacks import (
    EarlyStopping
)


# ======================================================================
# 2. SETTINGS
# ======================================================================

warnings.filterwarnings(
    "ignore"
)

SEED = 42

np.random.seed(
    SEED
)

random.seed(
    SEED
)

tf.random.set_seed(
    SEED
)


# ======================================================================
# 3. PROJECT DIRECTORY
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Rainfall Pattern Analysis and Prediction"
)


DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_june_2019_to_may_2020.csv"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ======================================================================
# 4. AIS OUTPUT PATHS
# ======================================================================

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "ais_rainfall_prediction_model.h5"
)


PKL_PATH = os.path.join(
    PROJECT_DIR,
    "ais_rainfall_preprocessor.pkl"
)


YAML_PATH = os.path.join(
    PROJECT_DIR,
    "ais_rainfall_config.yaml"
)


JSON_PATH = os.path.join(
    PROJECT_DIR,
    "ais_rainfall_metadata.json"
)


ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_accuracy_graph.png"
)


HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "ais_heatmap.png"
)


COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_comparison_graph.png"
)


RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_result.csv"
)


RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_result_graph.png"
)


PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_prediction.csv"
)


PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_prediction_graph.png"
)


AIS_HISTORY_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_optimization_history.csv"
)


AIS_OPTIMIZATION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_optimization_graph.png"
)


# ======================================================================
# 5. PROJECT INFORMATION
# ======================================================================

print(
    "=" * 78
)

print(
    "AI-BASED RAINFALL PATTERN ANALYSIS AND PREDICTION"
)

print(
    "ARTIFICIAL IMMUNE SYSTEM (AIS) OPTIMIZATION"
)

print(
    "=" * 78
)


print(
    "\nProject Directory:"
)

print(
    PROJECT_DIR
)


print(
    "\nDataset:"
)

print(
    DATASET_PATH
)


# ======================================================================
# 6. CHECK DATASET
# ======================================================================

if not os.path.exists(
    DATASET_PATH
):

    raise FileNotFoundError(
        f"\nDataset not found:\n"
        f"{DATASET_PATH}"
    )


print(
    "\nDataset found successfully."
)


# ======================================================================
# 7. LOAD DATASET
# ======================================================================

df = pd.read_csv(
    DATASET_PATH
)


df.columns = (
    df.columns
    .str.strip()
)


print(
    "\n" + "=" * 78
)

print(
    "DATASET INFORMATION"
)

print(
    "=" * 78
)


print(
    "\nDataset Shape:",
    df.shape
)


print(
    "\nColumns:"
)


for column in df.columns:

    print(
        "-",
        column
    )


print(
    "\nOriginal Dataset:"
)


print(
    df.to_string(
        index=False
    )
)


# ======================================================================
# 8. IDENTIFY COLUMNS
# ======================================================================

description_column = None

rainfall_column = None


for column in df.columns:

    lower_column = (
        column.lower()
    )

    if "description" in lower_column:

        description_column = column

    if "rainfall" in lower_column:

        rainfall_column = column


# Description fallback
if description_column is None:

    description_column = (
        df.columns[0]
    )


# Rainfall fallback
if rainfall_column is None:

    numeric_candidates = []

    for column in df.columns:

        converted = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        if (
            converted
            .notna()
            .sum()
            > 0
        ):

            numeric_candidates.append(
                column
            )

    if len(
        numeric_candidates
    ) > 0:

        rainfall_column = (
            numeric_candidates[0]
        )


if rainfall_column is None:

    raise ValueError(
        "Rainfall column could not be identified."
    )


print(
    "\nDescription Column:",
    description_column
)


print(
    "Rainfall Column:",
    rainfall_column
)


# ======================================================================
# 9. CLEAN DATA
# ======================================================================

df[
    description_column
] = (
    df[
        description_column
    ]
    .astype(str)
    .str.strip()
)


df[
    rainfall_column
] = pd.to_numeric(
    df[
        rainfall_column
    ],
    errors="coerce"
)


df = (
    df
    .dropna(
        subset=[
            rainfall_column
        ]
    )
    .reset_index(
        drop=True
    )
)


# ======================================================================
# 10. EXTRACT NORMAL RAINFALL
# ======================================================================

normal_rows = df[
    df[
        description_column
    ]
    .str.lower()
    .str.contains(
        "normal",
        na=False
    )
]


if len(
    normal_rows
) == 0:

    raise ValueError(
        "Normal Rainfall row was not found."
    )


normal_rainfall = float(
    normal_rows[
        rainfall_column
    ].iloc[0]
)


# ======================================================================
# 11. EXTRACT ACTUAL RAINFALL
# ======================================================================

actual_rows = df[
    df[
        description_column
    ]
    .str.lower()
    .str.contains(
        "actual",
        na=False
    )
]


if len(
    actual_rows
) == 0:

    raise ValueError(
        "Actual Rainfall row was not found."
    )


actual_rainfall = float(
    actual_rows[
        rainfall_column
    ].iloc[0]
)


# ======================================================================
# 12. RAINFALL ANALYSIS
# ======================================================================

difference = (
    actual_rainfall
    -
    normal_rainfall
)


percentage_deviation = (
    difference
    /
    normal_rainfall
) * 100


rainfall_ratio = (
    actual_rainfall
    /
    normal_rainfall
)


if percentage_deviation > 0:

    rainfall_pattern = (
        "Above Normal"
    )

elif percentage_deviation < 0:

    rainfall_pattern = (
        "Below Normal"
    )

else:

    rainfall_pattern = (
        "Normal"
    )


print(
    "\n" + "=" * 78
)

print(
    "RAINFALL ANALYSIS"
)

print(
    "=" * 78
)


print(
    f"\nNormal Rainfall      : "
    f"{normal_rainfall:.2f} mm"
)


print(
    f"Actual Rainfall      : "
    f"{actual_rainfall:.2f} mm"
)


print(
    f"Difference           : "
    f"{difference:.2f} mm"
)


print(
    f"Percentage Deviation : "
    f"{percentage_deviation:.4f}%"
)


print(
    f"Actual/Normal Ratio  : "
    f"{rainfall_ratio:.6f}"
)


print(
    f"Rainfall Pattern     : "
    f"{rainfall_pattern}"
)


# ======================================================================
# 13. CREATE SYNTHETIC TRAINING DATA
# ======================================================================
#
# IMPORTANT:
#
# The original supplied dataset contains only an aggregate Normal
# Rainfall and Actual Rainfall pair.
#
# This is insufficient for training and evaluating a meaningful ML
# model directly.
#
# Therefore, a local synthetic training dataset is generated around
# the observed relationship. AIS then optimizes the ANN trained on
# this demonstration dataset.
#
# ======================================================================

TOTAL_SAMPLES = 500


training_normal_values = np.linspace(

    normal_rainfall * 0.60,

    normal_rainfall * 1.40,

    TOTAL_SAMPLES
)


variation = (

    1.0

    +

    0.015

    *

    np.sin(

        np.linspace(
            0,
            8 * np.pi,
            TOTAL_SAMPLES
        )
    )
)


training_actual_values = (

    training_normal_values

    *

    rainfall_ratio

    *

    variation
)


X = (
    training_normal_values
    .reshape(
        -1,
        1
    )
)


y = (
    training_actual_values
    .reshape(
        -1,
        1
    )
)


print(
    "\nGenerated Training Samples:",
    len(X)
)


# ======================================================================
# 14. TRAIN / TEST SPLIT
# ======================================================================

(
    X_train,
    X_test,
    y_train,
    y_test

) = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=SEED
)


print(
    "Training Samples:",
    len(X_train)
)


print(
    "Testing Samples:",
    len(X_test)
)


# ======================================================================
# 15. SCALE DATA
# ======================================================================

feature_scaler = (
    StandardScaler()
)


target_scaler = (
    StandardScaler()
)


X_train_scaled = (
    feature_scaler
    .fit_transform(
        X_train
    )
)


X_test_scaled = (
    feature_scaler
    .transform(
        X_test
    )
)


y_train_scaled = (
    target_scaler
    .fit_transform(
        y_train
    )
)


y_test_scaled = (
    target_scaler
    .transform(
        y_test
    )
)


# ======================================================================
# 16. CREATE AIS TRAINING / VALIDATION SPLIT
# ======================================================================

(
    X_ais_train,
    X_ais_validation,
    y_ais_train,
    y_ais_validation

) = train_test_split(

    X_train_scaled,
    y_train_scaled,

    test_size=0.20,

    random_state=SEED
)


# ======================================================================
# 17. ANN CREATION FUNCTION
# ======================================================================

def create_ann_model(
    neurons_1,
    neurons_2,
    learning_rate
):

    """
    Build an ANN using hyperparameters supplied by AIS.
    """

    model = Sequential(
        [

            Input(
                shape=(1,)
            ),

            Dense(
                int(
                    neurons_1
                ),
                activation="relu"
            ),

            Dense(
                int(
                    neurons_2
                ),
                activation="relu"
            ),

            Dense(
                8,
                activation="relu"
            ),

            Dense(
                1,
                activation="linear"
            )

        ]
    )


    # --------------------------------------------------------------
    # IMPORTANT H5 COMPATIBILITY FIX
    # --------------------------------------------------------------
    #
    # Use explicit serializable Keras objects rather than:
    #
    # loss="mse"
    #
    # --------------------------------------------------------------

    model.compile(

        optimizer=tf.keras.optimizers.Adam(
            learning_rate=float(
                learning_rate
            )
        ),

        loss=tf.keras.losses.MeanSquaredError(),

        metrics=[
            tf.keras.metrics.MeanAbsoluteError(
                name="mae"
            )
        ]
    )


    return model


# ======================================================================
# 18. AIS PARAMETER RANGES
# ======================================================================

NEURON_1_MIN = 16
NEURON_1_MAX = 96

NEURON_2_MIN = 8
NEURON_2_MAX = 64

LEARNING_RATE_MIN = 0.0003
LEARNING_RATE_MAX = 0.005

BATCH_OPTIONS = [
    8,
    16,
    32
]


# ======================================================================
# 19. CREATE RANDOM ANTIBODY
# ======================================================================

def create_antibody():

    """
    Each antibody represents one ANN hyperparameter combination.
    """

    return {

        "neurons_1":
            random.randint(
                NEURON_1_MIN,
                NEURON_1_MAX
            ),

        "neurons_2":
            random.randint(
                NEURON_2_MIN,
                NEURON_2_MAX
            ),

        "learning_rate":
            random.uniform(
                LEARNING_RATE_MIN,
                LEARNING_RATE_MAX
            ),

        "batch_size":
            random.choice(
                BATCH_OPTIONS
            ),

        "fitness":
            None,

        "validation_loss":
            None

    }


# ======================================================================
# 20. AIS FITNESS FUNCTION
# ======================================================================

def evaluate_antibody(
    antibody
):

    """
    Train a small ANN and calculate AIS affinity.

    Lower validation MSE is better.

    Fitness / affinity:
        1 / (validation_loss + epsilon)

    Higher fitness is therefore better.
    """

    tf.keras.backend.clear_session()

    gc.collect()


    model = create_ann_model(

        antibody[
            "neurons_1"
        ],

        antibody[
            "neurons_2"
        ],

        antibody[
            "learning_rate"
        ]
    )


    early_stop = EarlyStopping(

        monitor="val_loss",

        patience=10,

        restore_best_weights=True,

        verbose=0
    )


    history = model.fit(

        X_ais_train,

        y_ais_train,

        validation_data=(
            X_ais_validation,
            y_ais_validation
        ),

        epochs=80,

        batch_size=int(
            antibody[
                "batch_size"
            ]
        ),

        callbacks=[
            early_stop
        ],

        verbose=0
    )


    validation_loss = min(
        history.history[
            "val_loss"
        ]
    )


    fitness = (
        1.0
        /
        (
            float(
                validation_loss
            )
            +
            1e-8
        )
    )


    antibody[
        "validation_loss"
    ] = float(
        validation_loss
    )


    antibody[
        "fitness"
    ] = float(
        fitness
    )


    del model

    tf.keras.backend.clear_session()

    gc.collect()


    return antibody


# ======================================================================
# 21. AIS CLONING FUNCTION
# ======================================================================

def clone_antibody(
    antibody
):

    """
    Create a copy of an antibody.
    """

    return {

        "neurons_1":
            int(
                antibody[
                    "neurons_1"
                ]
            ),

        "neurons_2":
            int(
                antibody[
                    "neurons_2"
                ]
            ),

        "learning_rate":
            float(
                antibody[
                    "learning_rate"
                ]
            ),

        "batch_size":
            int(
                antibody[
                    "batch_size"
                ]
            ),

        "fitness":
            None,

        "validation_loss":
            None

    }


# ======================================================================
# 22. AIS HYPERMUTATION
# ======================================================================

def mutate_antibody(
    antibody,
    mutation_strength=1.0
):

    """
    AIS hypermutation.

    Better antibodies receive smaller mutations while weaker clones
    can explore a larger region of the hyperparameter space.
    """

    mutated = clone_antibody(
        antibody
    )


    neuron1_change = random.randint(
        -16,
        16
    )


    neuron2_change = random.randint(
        -12,
        12
    )


    mutated[
        "neurons_1"
    ] = int(
        np.clip(

            mutated[
                "neurons_1"
            ]
            +
            neuron1_change
            *
            mutation_strength,

            NEURON_1_MIN,

            NEURON_1_MAX
        )
    )


    mutated[
        "neurons_2"
    ] = int(
        np.clip(

            mutated[
                "neurons_2"
            ]
            +
            neuron2_change
            *
            mutation_strength,

            NEURON_2_MIN,

            NEURON_2_MAX
        )
    )


    learning_rate_change = random.uniform(
        -0.001,
        0.001
    )


    mutated[
        "learning_rate"
    ] = float(
        np.clip(

            mutated[
                "learning_rate"
            ]
            +
            learning_rate_change
            *
            mutation_strength,

            LEARNING_RATE_MIN,

            LEARNING_RATE_MAX
        )
    )


    if random.random() < 0.35:

        mutated[
            "batch_size"
        ] = random.choice(
            BATCH_OPTIONS
        )


    return mutated


# ======================================================================
# 23. AIS CONFIGURATION
# ======================================================================
#
# Values are intentionally moderate so that the code remains practical
# on a normal laptop.
#
# Increase them if you want a larger search.
#
# ======================================================================

AIS_POPULATION_SIZE = 6

AIS_GENERATIONS = 4

AIS_SELECTION_SIZE = 3

AIS_CLONES_PER_ANTIBODY = 2

AIS_RANDOM_REPLACEMENTS = 1


# ======================================================================
# 24. INITIAL AIS POPULATION
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "STARTING ARTIFICIAL IMMUNE SYSTEM OPTIMIZATION"
)

print(
    "=" * 78
)


population = []


for i in range(
    AIS_POPULATION_SIZE
):

    antibody = (
        create_antibody()
    )


    print(
        f"\nEvaluating Initial Antibody "
        f"{i + 1}/{AIS_POPULATION_SIZE}"
    )


    antibody = (
        evaluate_antibody(
            antibody
        )
    )


    population.append(
        antibody
    )


    print(
        f"Neurons 1      : "
        f"{antibody['neurons_1']}"
    )


    print(
        f"Neurons 2      : "
        f"{antibody['neurons_2']}"
    )


    print(
        f"Learning Rate  : "
        f"{antibody['learning_rate']:.6f}"
    )


    print(
        f"Batch Size     : "
        f"{antibody['batch_size']}"
    )


    print(
        f"Validation Loss: "
        f"{antibody['validation_loss']:.8f}"
    )


    print(
        f"Fitness        : "
        f"{antibody['fitness']:.6f}"
    )


# ======================================================================
# 25. AIS OPTIMIZATION HISTORY
# ======================================================================

ais_history = []


# ======================================================================
# 26. AIS GENERATIONS
# ======================================================================

for generation in range(
    AIS_GENERATIONS
):

    print(
        "\n" + "=" * 78
    )

    print(
        f"AIS GENERATION "
        f"{generation + 1}/{AIS_GENERATIONS}"
    )

    print(
        "=" * 78
    )


    # --------------------------------------------------------------
    # Sort antibodies by affinity
    # --------------------------------------------------------------

    population = sorted(

        population,

        key=lambda item:
            item[
                "fitness"
            ],

        reverse=True
    )


    # --------------------------------------------------------------
    # Select strongest antibodies
    # --------------------------------------------------------------

    selected = population[
        :AIS_SELECTION_SIZE
    ]


    # --------------------------------------------------------------
    # Clone selected antibodies
    # --------------------------------------------------------------

    clones = []


    for rank, antibody in enumerate(
        selected
    ):

        # Best antibody gets smaller mutations.
        mutation_strength = (
            0.5
            +
            (
                rank
                /
                max(
                    1,
                    AIS_SELECTION_SIZE - 1
                )
            )
        )


        for clone_number in range(
            AIS_CLONES_PER_ANTIBODY
        ):

            clone = mutate_antibody(

                antibody,

                mutation_strength=
                    mutation_strength
            )


            print(
                f"\nEvaluating Clone "
                f"{clone_number + 1} "
                f"from Selected Antibody "
                f"{rank + 1}"
            )


            clone = (
                evaluate_antibody(
                    clone
                )
            )


            clones.append(
                clone
            )


    # --------------------------------------------------------------
    # Add random antibodies for diversity
    # --------------------------------------------------------------

    random_antibodies = []


    for replacement in range(
        AIS_RANDOM_REPLACEMENTS
    ):

        new_antibody = (
            create_antibody()
        )


        print(
            "\nEvaluating Random Diversity Antibody"
        )


        new_antibody = (
            evaluate_antibody(
                new_antibody
            )
        )


        random_antibodies.append(
            new_antibody
        )


    # --------------------------------------------------------------
    # Combine
    # --------------------------------------------------------------

    combined_population = (

        population

        +

        clones

        +

        random_antibodies
    )


    # --------------------------------------------------------------
    # Select next population
    # --------------------------------------------------------------

    combined_population = sorted(

        combined_population,

        key=lambda item:
            item[
                "fitness"
            ],

        reverse=True
    )


    population = combined_population[
        :AIS_POPULATION_SIZE
    ]


    best = population[0]


    average_fitness = float(
        np.mean(
            [
                antibody[
                    "fitness"
                ]
                for antibody in population
            ]
        )
    )


    ais_history.append(
        {

            "Generation":
                generation + 1,

            "Best_Fitness":
                best[
                    "fitness"
                ],

            "Average_Fitness":
                average_fitness,

            "Best_Validation_Loss":
                best[
                    "validation_loss"
                ],

            "Best_Neurons_1":
                best[
                    "neurons_1"
                ],

            "Best_Neurons_2":
                best[
                    "neurons_2"
                ],

            "Best_Learning_Rate":
                best[
                    "learning_rate"
                ],

            "Best_Batch_Size":
                best[
                    "batch_size"
                ]

        }
    )


    print(
        "\nBEST ANTIBODY OF GENERATION"
    )


    print(
        f"Fitness        : "
        f"{best['fitness']:.6f}"
    )


    print(
        f"Validation Loss: "
        f"{best['validation_loss']:.8f}"
    )


    print(
        f"Neurons 1      : "
        f"{best['neurons_1']}"
    )


    print(
        f"Neurons 2      : "
        f"{best['neurons_2']}"
    )


    print(
        f"Learning Rate  : "
        f"{best['learning_rate']:.6f}"
    )


    print(
        f"Batch Size     : "
        f"{best['batch_size']}"
    )


# ======================================================================
# 27. BEST AIS SOLUTION
# ======================================================================

population = sorted(

    population,

    key=lambda item:
        item[
            "fitness"
        ],

    reverse=True
)


best_antibody = (
    population[0]
)


BEST_NEURONS_1 = int(
    best_antibody[
        "neurons_1"
    ]
)


BEST_NEURONS_2 = int(
    best_antibody[
        "neurons_2"
    ]
)


BEST_LEARNING_RATE = float(
    best_antibody[
        "learning_rate"
    ]
)


BEST_BATCH_SIZE = int(
    best_antibody[
        "batch_size"
    ]
)


print(
    "\n" + "=" * 78
)

print(
    "AIS OPTIMIZATION COMPLETED"
)

print(
    "=" * 78
)


print(
    f"\nBest First Hidden Layer : "
    f"{BEST_NEURONS_1}"
)


print(
    f"Best Second Hidden Layer: "
    f"{BEST_NEURONS_2}"
)


print(
    f"Best Learning Rate      : "
    f"{BEST_LEARNING_RATE:.8f}"
)


print(
    f"Best Batch Size         : "
    f"{BEST_BATCH_SIZE}"
)


print(
    f"Best AIS Fitness        : "
    f"{best_antibody['fitness']:.8f}"
)


print(
    f"Best Validation Loss    : "
    f"{best_antibody['validation_loss']:.8f}"
)


# ======================================================================
# 28. SAVE AIS OPTIMIZATION HISTORY CSV
# ======================================================================

ais_history_df = pd.DataFrame(
    ais_history
)


ais_history_df.to_csv(
    AIS_HISTORY_CSV_PATH,
    index=False
)


print(
    "\nAIS Optimization History Saved:"
)

print(
    AIS_HISTORY_CSV_PATH
)


# ======================================================================
# 29. AIS OPTIMIZATION GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    ais_history_df[
        "Generation"
    ],

    ais_history_df[
        "Best_Fitness"
    ],

    marker="o",

    linewidth=2,

    label="Best AIS Fitness"
)


plt.plot(

    ais_history_df[
        "Generation"
    ],

    ais_history_df[
        "Average_Fitness"
    ],

    marker="s",

    linewidth=2,

    label="Average AIS Fitness"
)


plt.xlabel(
    "AIS Generation"
)


plt.ylabel(
    "Fitness"
)


plt.title(
    "AIS Optimization Progress"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    AIS_OPTIMIZATION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


# ======================================================================
# 30. CLEAR AIS TEMPORARY MODELS
# ======================================================================

tf.keras.backend.clear_session()

gc.collect()


# ======================================================================
# 31. CREATE FINAL AIS-OPTIMIZED MODEL
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "TRAINING FINAL AIS-OPTIMIZED ANN"
)

print(
    "=" * 78
)


final_model = create_ann_model(

    BEST_NEURONS_1,

    BEST_NEURONS_2,

    BEST_LEARNING_RATE
)


final_model.summary()


# ======================================================================
# 32. FINAL MODEL EARLY STOPPING
# ======================================================================

final_early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=40,

    restore_best_weights=True,

    verbose=0
)


# ======================================================================
# 33. TRAIN FINAL MODEL
# ======================================================================

final_history = final_model.fit(

    X_train_scaled,

    y_train_scaled,

    validation_split=0.20,

    epochs=500,

    batch_size=BEST_BATCH_SIZE,

    callbacks=[
        final_early_stopping
    ],

    verbose=0
)


epochs_completed = len(
    final_history.history[
        "loss"
    ]
)


print(
    "\nFinal AIS model training completed."
)


print(
    "Epochs Completed:",
    epochs_completed
)


# ======================================================================
# 34. TEST PREDICTIONS
# ======================================================================

test_prediction_scaled = (
    final_model.predict(
        X_test_scaled,
        verbose=0
    )
)


test_predictions = (
    target_scaler
    .inverse_transform(
        test_prediction_scaled
    )
)


actual_test_values = (
    y_test.flatten()
)


predicted_test_values = (
    test_predictions.flatten()
)


# ======================================================================
# 35. FINAL AIS MODEL METRICS
# ======================================================================

mae = mean_absolute_error(

    actual_test_values,

    predicted_test_values
)


mse = mean_squared_error(

    actual_test_values,

    predicted_test_values
)


rmse = np.sqrt(
    mse
)


r2 = r2_score(

    actual_test_values,

    predicted_test_values
)


safe_actual = np.where(

    actual_test_values == 0,

    1e-8,

    actual_test_values
)


mape = (

    np.mean(

        np.abs(

            (
                actual_test_values
                -
                predicted_test_values
            )

            /

            safe_actual
        )
    )

    *

    100
)


regression_accuracy = max(

    0.0,

    100.0
    -
    mape
)


print(
    "\n" + "=" * 78
)

print(
    "AIS MODEL PERFORMANCE"
)

print(
    "=" * 78
)


print(
    f"\nMAE                 : "
    f"{mae:.6f}"
)


print(
    f"MSE                 : "
    f"{mse:.6f}"
)


print(
    f"RMSE                : "
    f"{rmse:.6f}"
)


print(
    f"R2 Score            : "
    f"{r2:.6f}"
)


print(
    f"MAPE                : "
    f"{mape:.6f}%"
)


print(
    f"Regression Accuracy : "
    f"{regression_accuracy:.6f}%"
)


# ======================================================================
# 36. ORIGINAL DATA PREDICTION
# ======================================================================

original_input = np.array(
    [
        [
            normal_rainfall
        ]
    ],
    dtype=np.float32
)


original_scaled = (
    feature_scaler
    .transform(
        original_input
    )
)


original_prediction_scaled = (
    final_model.predict(
        original_scaled,
        verbose=0
    )
)


predicted_actual_rainfall = float(

    target_scaler
    .inverse_transform(
        original_prediction_scaled
    )[0][0]
)


prediction_error = abs(

    actual_rainfall

    -

    predicted_actual_rainfall
)


percentage_error = (

    prediction_error

    /

    actual_rainfall

) * 100


print(
    "\n" + "=" * 78
)

print(
    "AIS ORIGINAL RAINFALL PREDICTION"
)

print(
    "=" * 78
)


print(
    f"\nNormal Rainfall    : "
    f"{normal_rainfall:.2f} mm"
)


print(
    f"Actual Rainfall    : "
    f"{actual_rainfall:.2f} mm"
)


print(
    f"Predicted Rainfall : "
    f"{predicted_actual_rainfall:.2f} mm"
)


print(
    f"Absolute Error     : "
    f"{prediction_error:.6f} mm"
)


print(
    f"Percentage Error   : "
    f"{percentage_error:.6f}%"
)


# ======================================================================
# 37. SAVE AIS MODEL AS H5
# ======================================================================

final_model.save(
    MODEL_PATH
)


print(
    "\nAIS H5 Model Saved:"
)

print(
    MODEL_PATH
)


# ======================================================================
# 38. SAVE AIS PREPROCESSOR PKL
# ======================================================================

preprocessor_data = {

    "feature_scaler":
        feature_scaler,

    "target_scaler":
        target_scaler,

    "feature_name":
        "Normal Rainfall",

    "target_name":
        "Actual Rainfall",

    "description_column":
        description_column,

    "rainfall_column":
        rainfall_column,

    "normal_rainfall":
        normal_rainfall,

    "actual_rainfall":
        actual_rainfall,

    "rainfall_ratio":
        rainfall_ratio,

    "optimization":
        "Artificial Immune System",

    "ais_algorithm":
        "Clonal Selection",

    "best_neurons_1":
        BEST_NEURONS_1,

    "best_neurons_2":
        BEST_NEURONS_2,

    "best_learning_rate":
        BEST_LEARNING_RATE,

    "best_batch_size":
        BEST_BATCH_SIZE

}


joblib.dump(
    preprocessor_data,
    PKL_PATH
)


print(
    "\nAIS PKL Saved:"
)

print(
    PKL_PATH
)


# ======================================================================
# 39. AIS ACCURACY GRAPH
# ======================================================================

training_loss = np.array(
    final_history.history[
        "loss"
    ]
)


validation_loss = np.array(
    final_history.history[
        "val_loss"
    ]
)


training_score = (

    100.0

    /

    (
        1.0
        +
        training_loss
    )
)


validation_score = (

    100.0

    /

    (
        1.0
        +
        validation_loss
    )
)


epoch_numbers = np.arange(

    1,

    epochs_completed + 1
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    epoch_numbers,

    training_score,

    linewidth=2,

    label="AIS Training Score"
)


plt.plot(

    epoch_numbers,

    validation_score,

    linewidth=2,

    label="AIS Validation Score"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Learning Score (%)"
)


plt.title(
    "AIS-Optimized Rainfall Model Accuracy"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nAIS Accuracy Graph Saved:"
)

print(
    ACCURACY_GRAPH_PATH
)


# ======================================================================
# 40. CREATE DATA FOR AIS HEATMAP
# ======================================================================

correlation_df = pd.DataFrame(
    {

        "Normal Rainfall":
            training_normal_values,

        "Actual Rainfall":
            training_actual_values,

        "Difference":
            (
                training_actual_values
                -
                training_normal_values
            ),

        "Ratio":
            (
                training_actual_values
                /
                training_normal_values
            ),

        "Deviation":
            (
                (
                    training_actual_values
                    -
                    training_normal_values
                )

                /

                training_normal_values
            )
            * 100

    }
)


correlation_matrix = (
    correlation_df
    .corr()
)


# ======================================================================
# 41. AIS HEATMAP
# ======================================================================

plt.figure(
    figsize=(10, 8)
)


heatmap_image = plt.imshow(

    correlation_matrix,

    cmap="coolwarm",

    interpolation="nearest",

    vmin=-1,

    vmax=1
)


plt.colorbar(
    heatmap_image,
    label="Correlation"
)


heatmap_labels = (
    correlation_matrix
    .columns
    .tolist()
)


plt.xticks(

    range(
        len(
            heatmap_labels
        )
    ),

    heatmap_labels,

    rotation=45,

    ha="right"
)


plt.yticks(

    range(
        len(
            heatmap_labels
        )
    ),

    heatmap_labels
)


for i in range(
    len(
        heatmap_labels
    )
):

    for j in range(
        len(
            heatmap_labels
        )
    ):

        value = (
            correlation_matrix
            .iloc[
                i,
                j
            ]
        )


        plt.text(

            j,

            i,

            f"{value:.2f}",

            ha="center",

            va="center"
        )


plt.title(
    "AIS Rainfall Feature Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nAIS Heatmap Saved:"
)

print(
    HEATMAP_PATH
)


# ======================================================================
# 42. AIS COMPARISON GRAPH
# ======================================================================

comparison_labels = [

    "Regression Accuracy",

    "R2 Score",

    "MAPE"
]


comparison_values = [

    regression_accuracy,

    max(
        0.0,
        r2 * 100.0
    ),

    mape
]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(

    comparison_labels,

    comparison_values
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "AIS Rainfall Model Performance Comparison"
)


for bar, value in zip(
    bars,
    comparison_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        bar.get_height(),

        f"{value:.2f}",

        ha="center",

        va="bottom",

        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nAIS Comparison Graph Saved:"
)

print(
    COMPARISON_GRAPH_PATH
)


# ======================================================================
# 43. CREATE AIS RESULT CSV
# ======================================================================

result_df = pd.DataFrame(
    {

        "Normal_Rainfall_mm":
            X_test.flatten(),

        "Actual_Rainfall_mm":
            actual_test_values,

        "AIS_Predicted_Rainfall_mm":
            predicted_test_values

    }
)


result_df[
    "AIS_Absolute_Error_mm"
] = np.abs(

    result_df[
        "Actual_Rainfall_mm"
    ]

    -

    result_df[
        "AIS_Predicted_Rainfall_mm"
    ]
)


result_df[
    "AIS_Percentage_Error"
] = (

    result_df[
        "AIS_Absolute_Error_mm"
    ]

    /

    result_df[
        "Actual_Rainfall_mm"
    ]

) * 100


result_df = (

    result_df

    .sort_values(
        by="Normal_Rainfall_mm"
    )

    .reset_index(
        drop=True
    )
)


result_df.insert(

    0,

    "Sample_ID",

    np.arange(
        1,
        len(
            result_df
        )
        + 1
    )
)


result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)


print(
    "\nAIS Result CSV Saved:"
)

print(
    RESULT_CSV_PATH
)


# ======================================================================
# 44. AIS RESULT GRAPH
# ======================================================================

plt.figure(
    figsize=(12, 6)
)


plt.plot(

    result_df[
        "Sample_ID"
    ],

    result_df[
        "Actual_Rainfall_mm"
    ],

    linewidth=2,

    label="Actual Rainfall"
)


plt.plot(

    result_df[
        "Sample_ID"
    ],

    result_df[
        "AIS_Predicted_Rainfall_mm"
    ],

    linestyle="--",

    linewidth=2,

    label="AIS Predicted Rainfall"
)


plt.xlabel(
    "Test Sample"
)


plt.ylabel(
    "Rainfall (mm)"
)


plt.title(
    "AIS Actual vs Predicted Rainfall"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nAIS Result Graph Saved:"
)

print(
    RESULT_GRAPH_PATH
)


# ======================================================================
# 45. GENERATE AIS PREDICTION RANGE
# ======================================================================

prediction_normal_values = np.linspace(

    normal_rainfall * 0.70,

    normal_rainfall * 1.30,

    50
)


prediction_input = (
    prediction_normal_values
    .reshape(
        -1,
        1
    )
)


prediction_input_scaled = (
    feature_scaler
    .transform(
        prediction_input
    )
)


prediction_scaled = (
    final_model.predict(
        prediction_input_scaled,
        verbose=0
    )
)


prediction_values = (

    target_scaler

    .inverse_transform(
        prediction_scaled
    )

    .flatten()
)


# ======================================================================
# 46. AIS PREDICTION CSV
# ======================================================================

prediction_df = pd.DataFrame(
    {

        "Prediction_ID":
            np.arange(
                1,
                len(
                    prediction_normal_values
                )
                + 1
            ),

        "Normal_Rainfall_mm":
            prediction_normal_values,

        "AIS_Predicted_Actual_Rainfall_mm":
            prediction_values

    }
)


prediction_df[
    "AIS_Predicted_Difference_mm"
] = (

    prediction_df[
        "AIS_Predicted_Actual_Rainfall_mm"
    ]

    -

    prediction_df[
        "Normal_Rainfall_mm"
    ]
)


prediction_df[
    "AIS_Predicted_Deviation_Percentage"
] = (

    prediction_df[
        "AIS_Predicted_Difference_mm"
    ]

    /

    prediction_df[
        "Normal_Rainfall_mm"
    ]

) * 100


prediction_df[
    "AIS_Rainfall_Category"
] = np.where(

    prediction_df[
        "AIS_Predicted_Deviation_Percentage"
    ] > 0,

    "Above Normal",

    np.where(

        prediction_df[
            "AIS_Predicted_Deviation_Percentage"
        ] < 0,

        "Below Normal",

        "Normal"
    )
)


prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)


print(
    "\nAIS Prediction CSV Saved:"
)

print(
    PREDICTION_CSV_PATH
)


# ======================================================================
# 47. AIS PREDICTION GRAPH
# ======================================================================

plt.figure(
    figsize=(11, 6)
)


plt.plot(

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    prediction_df[
        "AIS_Predicted_Actual_Rainfall_mm"
    ],

    marker="o",

    markersize=4,

    linewidth=2,

    label="AIS Predicted Rainfall"
)


plt.plot(

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    prediction_df[
        "Normal_Rainfall_mm"
    ],

    linestyle="--",

    linewidth=2,

    label="Normal Rainfall Reference"
)


plt.scatter(

    [
        normal_rainfall
    ],

    [
        actual_rainfall
    ],

    s=100,

    label="Observed Rainfall",

    zorder=5
)


plt.xlabel(
    "Normal Rainfall (mm)"
)


plt.ylabel(
    "Rainfall (mm)"
)


plt.title(
    "AIS Rainfall Prediction Graph"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "\nAIS Prediction Graph Saved:"
)

print(
    PREDICTION_GRAPH_PATH
)


# ======================================================================
# 48. SAVE AIS YAML CONFIGURATION
# ======================================================================

yaml_config = {

    "project": {

        "name":
            "AI-Based Rainfall Pattern Analysis and Prediction",

        "optimization":
            "Artificial Immune System",

        "ais_algorithm":
            "Clonal Selection Algorithm",

        "version":
            "1.0"

    },

    "dataset": {

        "filename":
            os.path.basename(
                DATASET_PATH
            ),

        "path":
            DATASET_PATH,

        "rows":
            int(
                df.shape[0]
            ),

        "columns":
            int(
                df.shape[1]
            )

    },

    "rainfall_analysis": {

        "normal_rainfall_mm":
            float(
                normal_rainfall
            ),

        "actual_rainfall_mm":
            float(
                actual_rainfall
            ),

        "difference_mm":
            float(
                difference
            ),

        "percentage_deviation":
            float(
                percentage_deviation
            ),

        "pattern":
            rainfall_pattern

    },

    "ais": {

        "population_size":
            AIS_POPULATION_SIZE,

        "generations":
            AIS_GENERATIONS,

        "selection_size":
            AIS_SELECTION_SIZE,

        "clones_per_antibody":
            AIS_CLONES_PER_ANTIBODY,

        "best_neurons_1":
            BEST_NEURONS_1,

        "best_neurons_2":
            BEST_NEURONS_2,

        "best_learning_rate":
            float(
                BEST_LEARNING_RATE
            ),

        "best_batch_size":
            BEST_BATCH_SIZE,

        "best_fitness":
            float(
                best_antibody[
                    "fitness"
                ]
            )

    },

    "model_performance": {

        "mae":
            float(
                mae
            ),

        "mse":
            float(
                mse
            ),

        "rmse":
            float(
                rmse
            ),

        "r2_score":
            float(
                r2
            ),

        "mape":
            float(
                mape
            ),

        "regression_accuracy":
            float(
                regression_accuracy
            )

    }

}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as yaml_file:

    yaml.safe_dump(

        yaml_config,

        yaml_file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "\nAIS YAML Saved:"
)

print(
    YAML_PATH
)


# ======================================================================
# 49. SAVE AIS JSON METADATA
# ======================================================================

json_metadata = {

    "project_name":
        "AI-Based Rainfall Pattern Analysis and Prediction",

    "optimization_algorithm":
        "Artificial Immune System",

    "ais_method":
        "Clonal Selection Algorithm",

    "dataset":
        os.path.basename(
            DATASET_PATH
        ),

    "rainfall_statistics": {

        "normal_rainfall_mm":
            round(
                normal_rainfall,
                4
            ),

        "actual_rainfall_mm":
            round(
                actual_rainfall,
                4
            ),

        "difference_mm":
            round(
                difference,
                4
            ),

        "deviation_percentage":
            round(
                percentage_deviation,
                4
            ),

        "pattern":
            rainfall_pattern

    },

    "best_ais_parameters": {

        "neurons_1":
            BEST_NEURONS_1,

        "neurons_2":
            BEST_NEURONS_2,

        "learning_rate":
            BEST_LEARNING_RATE,

        "batch_size":
            BEST_BATCH_SIZE,

        "fitness":
            best_antibody[
                "fitness"
            ],

        "validation_loss":
            best_antibody[
                "validation_loss"
            ]

    },

    "model_performance": {

        "mae":
            round(
                mae,
                6
            ),

        "mse":
            round(
                mse,
                6
            ),

        "rmse":
            round(
                rmse,
                6
            ),

        "r2":
            round(
                r2,
                6
            ),

        "mape":
            round(
                mape,
                6
            ),

        "regression_accuracy":
            round(
                regression_accuracy,
                6
            )

    },

    "original_prediction": {

        "normal_rainfall_mm":
            round(
                normal_rainfall,
                4
            ),

        "actual_rainfall_mm":
            round(
                actual_rainfall,
                4
            ),

        "ais_predicted_rainfall_mm":
            round(
                predicted_actual_rainfall,
                4
            ),

        "absolute_error_mm":
            round(
                prediction_error,
                6
            ),

        "percentage_error":
            round(
                percentage_error,
                6
            )

    },

    "generated_files": [

        "ais_rainfall_prediction_model.h5",

        "ais_rainfall_preprocessor.pkl",

        "ais_rainfall_config.yaml",

        "ais_rainfall_metadata.json",

        "ais_accuracy_graph.png",

        "ais_heatmap.png",

        "ais_comparison_graph.png",

        "ais_result.csv",

        "ais_result_graph.png",

        "ais_prediction.csv",

        "ais_prediction_graph.png",

        "ais_optimization_history.csv",

        "ais_optimization_graph.png"

    ]

}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as json_file:

    json.dump(

        json_metadata,

        json_file,

        indent=4,

        ensure_ascii=False
    )


print(
    "\nAIS JSON Saved:"
)

print(
    JSON_PATH
)


# ======================================================================
# 50. VERIFY SAVED AIS H5 MODEL
# ======================================================================
#
# IMPORTANT:
#
# compile=False avoids the Keras legacy H5:
#
# TypeError:
# Could not locate function 'mse'
#
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "VERIFYING SAVED AIS MODEL"
)

print(
    "=" * 78
)


loaded_model = (
    tf.keras.models.load_model(
        MODEL_PATH,
        compile=False
    )
)


loaded_preprocessor = (
    joblib.load(
        PKL_PATH
    )
)


loaded_feature_scaler = (
    loaded_preprocessor[
        "feature_scaler"
    ]
)


loaded_target_scaler = (
    loaded_preprocessor[
        "target_scaler"
    ]
)


verify_input = np.array(
    [
        [
            normal_rainfall
        ]
    ],
    dtype=np.float32
)


verify_scaled = (
    loaded_feature_scaler
    .transform(
        verify_input
    )
)


verify_prediction_scaled = (
    loaded_model.predict(
        verify_scaled,
        verbose=0
    )
)


verify_prediction = float(

    loaded_target_scaler

    .inverse_transform(
        verify_prediction_scaled
    )[0][0]
)


print(
    "\nAIS H5 Model Loaded Successfully."
)


print(
    f"Verification Prediction: "
    f"{verify_prediction:.2f} mm"
)


# ======================================================================
# 51. AIS CUSTOM PREDICTION FUNCTION
# ======================================================================

def ais_predict_actual_rainfall(
    normal_rainfall_input
):

    """
    Load the saved AIS-optimized ANN and predict actual rainfall.
    """

    model_loaded = (
        tf.keras.models.load_model(
            MODEL_PATH,
            compile=False
        )
    )


    preprocessing_loaded = (
        joblib.load(
            PKL_PATH
        )
    )


    x_scaler = (
        preprocessing_loaded[
            "feature_scaler"
        ]
    )


    y_scaler = (
        preprocessing_loaded[
            "target_scaler"
        ]
    )


    input_data = np.array(
        [
            [
                float(
                    normal_rainfall_input
                )
            ]
        ],
        dtype=np.float32
    )


    input_scaled = (
        x_scaler
        .transform(
            input_data
        )
    )


    prediction_scaled = (
        model_loaded.predict(
            input_scaled,
            verbose=0
        )
    )


    prediction = (
        y_scaler
        .inverse_transform(
            prediction_scaled
        )
    )


    return float(
        prediction[0][0]
    )


# ======================================================================
# 52. EXAMPLE AIS PREDICTION
# ======================================================================

example_normal_rainfall = (
    1000.0
)


example_prediction = (
    ais_predict_actual_rainfall(
        example_normal_rainfall
    )
)


print(
    "\n" + "=" * 78
)

print(
    "AIS CUSTOM PREDICTION"
)

print(
    "=" * 78
)


print(
    f"\nNormal Rainfall Input:"
    f" {example_normal_rainfall:.2f} mm"
)


print(
    f"AIS Predicted Rainfall:"
    f" {example_prediction:.2f} mm"
)


# ======================================================================
# 53. RESULT CSV PREVIEW
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "AIS RESULT CSV PREVIEW"
)

print(
    "=" * 78
)


print(

    result_df

    .head(
        10
    )

    .to_string(
        index=False
    )
)


# ======================================================================
# 54. PREDICTION CSV PREVIEW
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "AIS PREDICTION CSV PREVIEW"
)

print(
    "=" * 78
)


print(

    prediction_df

    .head(
        10
    )

    .to_string(
        index=False
    )
)


# ======================================================================
# 55. VERIFY ALL AIS FILES
# ======================================================================

all_ais_files = [

    MODEL_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    AIS_HISTORY_CSV_PATH,

    AIS_OPTIMIZATION_GRAPH_PATH

]


print(
    "\n" + "=" * 78
)

print(
    "VERIFYING ALL AIS OUTPUT FILES"
)

print(
    "=" * 78
)


for file_path in all_ais_files:

    if os.path.exists(
        file_path
    ):

        file_size_kb = (

            os.path.getsize(
                file_path
            )

            /

            1024
        )


        print(

            f"[OK] "

            f"{os.path.basename(file_path)} "

            f"({file_size_kb:.2f} KB)"
        )

    else:

        print(

            f"[ERROR] "

            f"{os.path.basename(file_path)}"
        )


# ======================================================================
# 56. FINAL SUMMARY
# ======================================================================

print(
    "\n" + "=" * 78
)

print(
    "AIS PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 78
)


print(
    f"""
PROJECT
-------
AI-Based Rainfall Pattern Analysis and Prediction


OPTIMIZATION
------------
Artificial Immune System (AIS)

AIS Method:
Clonal Selection Algorithm


DATASET
-------
{DATASET_PATH}


RAINFALL ANALYSIS
-----------------
Normal Rainfall:
{normal_rainfall:.2f} mm

Actual Rainfall:
{actual_rainfall:.2f} mm

Difference:
{difference:.2f} mm

Deviation:
{percentage_deviation:.4f}%

Pattern:
{rainfall_pattern}


BEST AIS PARAMETERS
-------------------
First Hidden Layer:
{BEST_NEURONS_1}

Second Hidden Layer:
{BEST_NEURONS_2}

Learning Rate:
{BEST_LEARNING_RATE:.8f}

Batch Size:
{BEST_BATCH_SIZE}

AIS Fitness:
{best_antibody['fitness']:.8f}

Validation Loss:
{best_antibody['validation_loss']:.8f}


AIS MODEL PERFORMANCE
---------------------
MAE:
{mae:.6f}

MSE:
{mse:.6f}

RMSE:
{rmse:.6f}

R2 Score:
{r2:.6f}

MAPE:
{mape:.6f}%

Regression Accuracy:
{regression_accuracy:.6f}%


ORIGINAL DATA PREDICTION
------------------------
Normal Rainfall:
{normal_rainfall:.2f} mm

Actual Rainfall:
{actual_rainfall:.2f} mm

AIS Predicted Rainfall:
{predicted_actual_rainfall:.2f} mm


GENERATED AIS FILES
-------------------
1. ais_rainfall_prediction_model.h5

2. ais_rainfall_preprocessor.pkl

3. ais_rainfall_config.yaml

4. ais_rainfall_metadata.json

5. ais_accuracy_graph.png

6. ais_heatmap.png

7. ais_comparison_graph.png

8. ais_result.csv

9. ais_result_graph.png

10. ais_prediction.csv

11. ais_prediction_graph.png

12. ais_optimization_history.csv

13. ais_optimization_graph.png


H5 COMPATIBILITY
-----------------
The model is loaded using:

tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

This avoids the legacy Keras H5
'mse' deserialization error.


OUTPUT DIRECTORY
----------------
{PROJECT_DIR}
"""
)


print(
    "=" * 78
)
```
